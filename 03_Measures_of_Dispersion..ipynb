# 1. Range

## Step 1 – Understand the Concept

Range is a simple measure of dispersion that tells us how far apart the smallest and largest values are in a dataset.

It is calculated by subtracting the minimum value from the maximum value.

### Formula

Range = Maximum Value - Minimum Value

A small range means the values are closer together, while a large range means the values are more spread out.

## Step 2 – Demonstrate the Concept

### Numerical Example

Consider the following daily sales:

10, 15, 20, 25, 30

- Maximum value = 30
- Minimum value = 10
- Range = 30 - 10 = 20

So, the sales values have a range of 20 units.

### Real-World Example

A shop can use range to understand the difference between its highest and lowest daily sales.

### AI/ML Use Case

Range can be used during Exploratory Data Analysis (EDA) to understand the overall spread of a numerical feature.

```python
# Create a dataset of daily sales
sales = [10, 15, 20, 25, 30]

# Find the maximum and minimum values
maximum_value = max(sales)
minimum_value = min(sales)

# Calculate the range
data_range = maximum_value - minimum_value

print("Maximum value:", maximum_value)
print("Minimum value:", minimum_value)
print("Range:", data_range)
```

## Step 3 – Explanation

### What does the program do?

The program finds the maximum and minimum sales values and subtracts the minimum from the maximum to calculate the range.

### What statistical concept does it demonstrate?

It demonstrates the concept of Range, which measures the total spread between the smallest and largest values.

### How do we interpret the output?

The output shows:

- Maximum value = 30
- Minimum value = 10
- Range = 20

This means the difference between the highest and lowest sales is 20 units.

### Where is this used in AI/ML?

Range is useful during data exploration to understand the spread of numerical features before applying preprocessing or machine learning algorithms.

# 2. Variance

## Step 1 – Understand the Concept

Variance measures how much the values in a dataset differ from their mean.

It calculates the average of the squared differences between each value and the mean.

- Low variance means the values are close to the mean.
- High variance means the values are more spread out.

### Formula

Population Variance = Sum of squared differences from the mean / Number of values

Squaring the differences makes all the values positive before calculating the average.

## Step 2 – Demonstrate the Concept

### Numerical Example

Consider:

2, 4, 6

First, calculate the mean:

Mean = (2 + 4 + 6) / 3 = 4

Now calculate the squared differences from the mean:

- For 2: (2 - 4)² = 4
- For 4: (4 - 4)² = 0
- For 6: (6 - 4)² = 4

Variance = (4 + 0 + 4) / 3

Variance = 2.67 approximately

### Real-World Example

If daily sales have high variance, the sales values change significantly from one day to another.

### AI/ML Use Case

Variance is useful for understanding feature variability and is an important concept in techniques such as Principal Component Analysis (PCA).

```python
import numpy as np

# Create a dataset
data = np.array([2, 4, 6])

# Calculate the mean
mean_value = np.mean(data)

# Calculate population variance
variance = np.var(data)

print("Dataset:", data)
print("Mean:", mean_value)
print("Population Variance:", variance)
```

## Step 3 – Explanation

### What does the program do?

The program calculates the mean and population variance of the dataset.

### What statistical concept does it demonstrate?

It demonstrates variance, which measures the average squared distance of values from the mean.

### How do we interpret the output?

The variance is approximately 2.67.

This tells us that the values have some amount of spread around their mean of 4.

### Where is this used in AI/ML?

Variance is used in data analysis, feature analysis, and dimensionality reduction techniques such as PCA.

# 3. Standard Deviation

## Step 1 – Understand the Concept

Standard deviation tells us how much the values in a dataset typically differ from the mean.

It is calculated as the square root of variance.

- Small standard deviation means the values are close to the mean.
- Large standard deviation means the values are more spread out.

One advantage of standard deviation is that it is expressed in the same unit as the original data.

## Step 2 – Demonstrate the Concept

### Numerical Example

Consider two datasets:

Dataset A = 48, 50, 52

Dataset B = 20, 50, 80

Both datasets have a mean of 50.

However, Dataset B is more spread out from the mean.

Therefore:

- Dataset A has a smaller standard deviation.
- Dataset B has a larger standard deviation.

### Real-World Example

For employee salaries, standard deviation can help understand how much salaries vary around the average salary.

### AI/ML Use Case

Standard deviation is commonly used when standardizing numerical features before training machine learning models.

```python
import numpy as np

# Create two datasets
data_a = np.array([48, 50, 52])
data_b = np.array([20, 50, 80])

# Calculate standard deviation
std_a = np.std(data_a)
std_b = np.std(data_b)

print("Dataset A:", data_a)
print("Standard Deviation of A:", round(std_a, 2))

print("\nDataset B:", data_b)
print("Standard Deviation of B:", round(std_b, 2))
```

## Step 3 – Explanation

### What does the program do?

The program calculates the standard deviation of two datasets.

### What statistical concept does it demonstrate?

It demonstrates how standard deviation measures the spread of values around the mean.

### How do we interpret the output?

Dataset A has a smaller standard deviation because its values are close to 50.

Dataset B has a larger standard deviation because its values are farther away from 50.

### Where is this used in AI/ML?

Standard deviation is used in feature scaling, data analysis, and identifying how much numerical features vary.

# 4. Quartiles

## Step 1 – Understand the Concept

Quartiles divide an ordered dataset into four parts.

The three important quartiles are:

- Q1: First Quartile – 25% of the data is at or below this value.
- Q2: Second Quartile – 50% of the data is at or below this value. It is also the median.
- Q3: Third Quartile – 75% of the data is at or below this value.

Quartiles help us understand how the data is distributed.

## Step 2 – Demonstrate the Concept

### Numerical Example

Consider:

10, 20, 30, 40, 50, 60, 70

For this dataset:

- Q1 = 20
- Q2 = 40
- Q3 = 60

This means:

- Around 25% of the data is at or below 20.
- Around 50% of the data is at or below 40.
- Around 75% of the data is at or below 60.

### Real-World Example

A company can use salary quartiles to understand how employee salaries are distributed.

### AI/ML Use Case

Quartiles are useful during Exploratory Data Analysis and can help identify possible outliers.

```python
import numpy as np

# Create a dataset
data = np.array([10, 20, 30, 40, 50, 60, 70])

# Calculate quartiles
q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)

print("Dataset:", data)
print("Q1:", q1)
print("Q2 (Median):", q2)
print("Q3:", q3)
```

## Step 3 – Explanation

### What does the program do?

The program calculates the first quartile, median, and third quartile using NumPy.

### What statistical concept does it demonstrate?

It demonstrates how quartiles divide ordered data into different parts.

### How do we interpret the output?

- Q1 = 20
- Q2 = 40
- Q3 = 60

These values describe different positions within the dataset.

### Where is this used in AI/ML?

Quartiles are used during data exploration to understand distributions and identify unusual observations.

# 5. Percentiles

## Step 1 – Understand the Concept

A percentile tells us the value below which a particular percentage of observations falls.

For example, the 90th percentile is the value below which approximately 90% of the observations fall.

Some important relationships are:

- 25th percentile = Q1
- 50th percentile = Q2 = Median
- 75th percentile = Q3

Percentiles are useful for understanding the relative position of values in a dataset.

## Step 2 – Demonstrate the Concept

### Numerical Example

Consider:

10, 20, 30, 40, 50, 60, 70, 80, 90

We can calculate different percentiles such as the 25th, 50th, 80th, and 90th percentiles.

For example, the 50th percentile represents the middle of the dataset.

### Real-World Example

A company can use the 90th percentile of delivery time to understand a high-end delivery-time level.

### AI/ML Use Case

Percentiles can be used to understand prediction errors and define thresholds for identifying potentially unusual observations.

```python
import numpy as np

# Create a dataset
data = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90])

# Calculate selected percentiles
p25 = np.percentile(data, 25)
p50 = np.percentile(data, 50)
p80 = np.percentile(data, 80)
p90 = np.percentile(data, 90)

print("25th Percentile:", p25)
print("50th Percentile:", p50)
print("80th Percentile:", p80)
print("90th Percentile:", p90)
```

## Step 3 – Explanation

### What does the program do?

The program calculates the 25th, 50th, 80th, and 90th percentiles of the dataset.

### What statistical concept does it demonstrate?

It demonstrates how percentiles describe the position of values within a dataset.

### How do we interpret the output?

The 50th percentile represents the median.

The 90th percentile represents a value below which approximately 90% of the observations fall, according to the percentile calculation method used.

### Where is this used in AI/ML?

Percentiles can be used for data analysis, error analysis, threshold setting, and identifying potentially unusual observations.

# 6. Interquartile Range (IQR)

## Step 1 – Understand the Concept

Interquartile Range, or IQR, measures the spread of the middle 50% of a dataset.

It is calculated using the third quartile and first quartile.

### Formula

IQR = Q3 - Q1

IQR is useful because it focuses on the middle part of the data and is less affected by extreme values.

It is also commonly used for identifying potential outliers.

## Step 2 – Demonstrate the Concept

### Numerical Example

Consider:

10, 20, 30, 40, 50, 60, 70

From the previous example:

- Q1 = 20
- Q3 = 60

Therefore:

IQR = Q3 - Q1

IQR = 60 - 20

IQR = 40

So, the middle 50% of the dataset is spread across 40 units.

### Real-World Example

A company can use IQR to understand the central spread of employee salaries without being strongly affected by extremely high or low salaries.

### AI/ML Use Case

IQR is commonly used during data preprocessing to identify possible outliers.

```python
import numpy as np

# Create a dataset
data = np.array([10, 20, 30, 40, 50, 60, 70])

# Calculate Q1 and Q3
q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)

# Calculate IQR
iqr = q3 - q1

# Calculate outlier boundaries
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

# Find possible outliers
outliers = data[(data < lower_bound) | (data > upper_bound)]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Possible Outliers:", outliers)
```

## Step 3 – Explanation

### What does the program do?

The program calculates Q1, Q3, and IQR.

It then calculates the lower and upper boundaries using the IQR method and checks whether any values fall outside those boundaries.

### What statistical concept does it demonstrate?

It demonstrates how IQR measures the spread of the middle 50% of the data and how it can be used for potential outlier detection.

### How do we interpret the output?

For this dataset:

- Q1 = 20
- Q3 = 60
- IQR = 40

The lower and upper boundaries are also calculated.

There are no possible outliers in this dataset because all values fall within the boundaries.

### Where is this used in AI/ML?

IQR can be used during data cleaning and Exploratory Data Analysis to identify potentially unusual observations before model training.

# 7. Coefficient of Variation (CV)

## Step 1 – Understand the Concept

Coefficient of Variation (CV) measures the amount of variation relative to the mean.

It is expressed as a percentage.

### Formula

CV = (Standard Deviation / Mean) × 100

A higher CV means there is more variability relative to the mean.

CV is especially useful when comparing the relative variation of datasets with different scales.

## Step 2 – Demonstrate the Concept

### Numerical Example

Consider two datasets:

Dataset A = 90, 100, 110

Dataset B = 900, 1000, 1100

Dataset B has larger values and therefore has a larger standard deviation.

However, both datasets have the same relative spread compared with their means.

The Coefficient of Variation helps us compare this relative variability.

### Real-World Example

A company can compare the relative variation in sales between two products even when their average sales are very different.

### AI/ML Use Case

CV can be used to compare the relative variability of numerical features with different scales.

```python
import numpy as np

# Create two datasets
data_a = np.array([90, 100, 110])
data_b = np.array([900, 1000, 1100])

# Calculate means
mean_a = np.mean(data_a)
mean_b = np.mean(data_b)

# Calculate standard deviations
std_a = np.std(data_a)
std_b = np.std(data_b)

# Calculate coefficient of variation
cv_a = (std_a / mean_a) * 100
cv_b = (std_b / mean_b) * 100

print("Dataset A")
print("Mean:", mean_a)
print("Standard Deviation:", round(std_a, 2))
print("Coefficient of Variation:", round(cv_a, 2), "%")

print("\nDataset B")
print("Mean:", mean_b)
print("Standard Deviation:", round(std_b, 2))
print("Coefficient of Variation:", round(cv_b, 2), "%")
```

## Step 3 – Explanation

### What does the program do?

The program calculates the mean and standard deviation of two datasets.

It then uses these values to calculate the Coefficient of Variation for each dataset.

### What statistical concept does it demonstrate?

It demonstrates relative variability by comparing the standard deviation with the mean.

### How do we interpret the output?

Dataset A and Dataset B have different absolute values and different standard deviations.

However, their Coefficient of Variation is the same because their relative spread is the same.

### Where is this used in AI/ML?

CV can be used to compare the relative variability of numerical features, especially when their scales are different.

It should be used carefully when the mean is zero or very close to zero.
