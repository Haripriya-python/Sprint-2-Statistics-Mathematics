## Question 6 – Correlation and Covariance

### Question:

Explain Correlation and Covariance.

### Coding Exercise:

Write a Python program using NumPy to calculate the Correlation and Covariance between the following two variables:

X = [10, 20, 30, 40, 50]

Y = [15, 25, 35, 45, 55]

```python
import numpy as np

X = [10, 20, 30, 40, 50]
Y = [15, 25, 35, 45, 55]

correlation = np.corrcoef(X, Y)[0, 1]
covariance = np.cov(X, Y)[0, 1]

print("Correlation:", correlation)
print("Covariance:", covariance)
```

## Question 10 – Type I and Type II Error

### Question:

What is the difference between Type I Error and Type II Error?

### Coding Exercise:

Write a Python program to demonstrate Type I Error and Type II Error using hypothesis testing.

```python
from scipy.stats import ttest_1samp

data = [10, 12, 11, 13, 15]

t_statistic, p_value = ttest_1samp(data, 10)

alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")

print("P-value:", p_value)
```

## Question 12 – Outlier Detection using IQR

### Question:

Detect Outliers using the IQR Method.

### Coding Exercise:

Write a Python program using NumPy to detect outliers in the following dataset using the IQR Method:

[10, 12, 15, 18, 20, 22, 25, 100]

```python
import numpy as np

data = np.array([10, 12, 15, 18, 20, 22, 25, 100])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[(data < lower_bound) | (data > upper_bound)]

print("Outliers:", outliers)
```
