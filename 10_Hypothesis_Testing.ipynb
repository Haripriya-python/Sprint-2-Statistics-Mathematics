# Null Hypothesis

## Step 1 – Understand the Concept

The **Null Hypothesis (H₀)** is a statement that assumes there is **no significant difference, effect, or relationship** in the data.

It is the starting assumption in hypothesis testing.

### Simple Example

Suppose a company claims that the average delivery time is **30 minutes**.

We want to test whether the actual average delivery time is different from 30 minutes.

The Null Hypothesis can be:

**H₀: The average delivery time is 30 minutes.**

### Key Point

The null hypothesis represents the **default assumption** that we test using sample data.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company introduces a new training program and wants to know whether it changes employee productivity.

We start with:

**H₀: The training program does not change employee productivity.**

Then we collect sample data and perform a statistical test.

If the evidence is strong enough, we may reject the null hypothesis.

### AI/ML Usage

Hypothesis testing can be used to:

- Compare model performance
- Test whether a feature has a significant effect
- Compare different data groups
- Evaluate whether an observed difference is statistically significant

```python
import numpy as np

# Create sample delivery time data
delivery_times = np.array([29, 31, 30, 32, 28, 30, 31, 29, 30, 31])

# Calculate the sample mean
sample_mean = np.mean(delivery_times)

# Assumed population mean
assumed_mean = 30

print("Delivery Times:", delivery_times)
print("Sample Mean:", sample_mean)
print("Assumed Mean:", assumed_mean)

# Null Hypothesis
print("H₀: The average delivery time is 30 minutes.")
```

## Step 3 – Explanation

### What does the program do?

The program calculates the average delivery time from the sample data and compares it with the assumed population mean.

### What statistical concept does it demonstrate?

It demonstrates the idea of the **Null Hypothesis (H₀)**.

### How do we interpret the output?

The null hypothesis states that the average delivery time is **30 minutes**.

The sample data is used to determine whether there is enough statistical evidence to reject this assumption.

### Where is this used in AI/ML?

Null hypotheses are used when testing whether:

- A model improvement is significant
- A feature has an effect
- Two groups are significantly different
- An observed pattern is statistically meaningful

# Alternative Hypothesis

## Step 1 – Understand the Concept

The **Alternative Hypothesis (H₁ or Hₐ)** is a statement that suggests there **is a significant difference, effect, or relationship** in the data.

It is the opposite of the Null Hypothesis.

### Simple Example

A company claims that the average delivery time is **30 minutes**.

We want to check whether the actual average delivery time is different from 30 minutes.

### Hypotheses

**H₀:** The average delivery time is 30 minutes.

**H₁:** The average delivery time is not 30 minutes.

### Key Point

The alternative hypothesis represents what we are trying to find evidence for during hypothesis testing.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company introduces a new training program and wants to know whether it changes employee productivity.

### Hypotheses

**H₀:** The training program does not change employee productivity.

**H₁:** The training program changes employee productivity.

We collect sample data and perform a statistical test.

If the evidence is strong enough, we reject the Null Hypothesis and support the Alternative Hypothesis.

### AI/ML Usage

Alternative hypotheses can be used to:

- Compare model performance
- Test whether a feature has an effect
- Compare two groups
- Test whether a change produces a significant result

```python
import numpy as np

# Create sample delivery time data
delivery_times = np.array([29, 31, 30, 32, 28, 30, 31, 29, 30, 31])

# Calculate the sample mean
sample_mean = np.mean(delivery_times)

# Assumed population mean
assumed_mean = 30

print("Delivery Times:", delivery_times)
print("Sample Mean:", sample_mean)
print("Assumed Mean:", assumed_mean)

# Alternative Hypothesis
print("H₁: The average delivery time is not 30 minutes.")
```

## Step 3 – Explanation

### What does the program do?

The program calculates the sample mean and compares it with the assumed population mean.

### What statistical concept does it demonstrate?

It demonstrates the **Alternative Hypothesis (H₁)**.

### How do we interpret the output?

The Alternative Hypothesis states that the average delivery time is **not 30 minutes**.

Statistical testing is used to determine whether there is enough evidence to support this statement.

### Where is this used in AI/ML?

Alternative hypotheses are useful when testing:

- Model performance differences
- Feature effects
- Differences between datasets
- Changes in business or experimental results

# p-value

## Step 1 – Understand the Concept

The **p-value** helps us determine whether the observed result is statistically significant under the assumption that the Null Hypothesis is true.

### Simple Meaning

A **small p-value** means there is stronger evidence against the Null Hypothesis.

A **large p-value** means there is not enough evidence to reject the Null Hypothesis.

### Common Rule

Using a significance level of **0.05**:

- **p-value ≤ 0.05** → Reject H₀
- **p-value > 0.05** → Fail to reject H₀

### Important Point

A p-value does **not** tell us the probability that the Null Hypothesis is true.

It tells us how unusual the observed data would be if the Null Hypothesis were true.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company introduces a new training program.

We want to know whether the training program changes employee productivity.

### Hypotheses

**H₀:** The training program does not change productivity.

**H₁:** The training program changes productivity.

After performing a statistical test, suppose we get:

**p-value = 0.03**

Since:

**0.03 < 0.05**

we reject the Null Hypothesis.

This means the result provides statistical evidence of a difference.

### AI/ML Usage

p-values can be used to:

- Test whether a feature has a significant relationship with an outcome
- Compare groups
- Evaluate experimental results
- Support statistical analysis during data exploration

```python
from scipy import stats
import numpy as np

# Create sample data
group_a = np.array([72, 75, 78, 74, 77, 76, 73, 79])
group_b = np.array([65, 68, 70, 67, 66, 69, 64, 71])

# Perform an independent t-test
t_statistic, p_value = stats.ttest_ind(group_a, group_b)

print("T-Statistic:", t_statistic)
print("p-value:", p_value)

# Set significance level
alpha = 0.05

# Make a decision
if p_value <= alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

## Step 3 – Explanation

### What does the program do?

The program compares two groups using an independent t-test and calculates the p-value.

### What statistical concept does it demonstrate?

It demonstrates how the **p-value** is used in hypothesis testing.

### How do we interpret the output?

We compare the p-value with the significance level **0.05**.

- If p-value ≤ 0.05 → Reject H₀
- If p-value > 0.05 → Fail to reject H₀

### Where is this used in AI/ML?

p-values can be used during statistical analysis to determine whether observed differences or relationships provide enough evidence to reject the Null Hypothesis.

### Important Point

A small p-value indicates stronger evidence against H₀, but it does not by itself tell us that the effect is practically important.

# Confidence Interval

## Step 1 – Understand the Concept

A **Confidence Interval (CI)** is a range of values used to estimate an unknown population parameter from sample data.

It gives us a range in which the true population value is expected to lie, based on a chosen confidence level.

### Common Confidence Levels

- 90% Confidence Interval
- 95% Confidence Interval
- 99% Confidence Interval

### Simple Example

Suppose the average delivery time from a sample is **30 minutes**.

A 95% confidence interval might be:

**28 minutes to 32 minutes**

This means the interval provides a range of plausible values for the population mean.

### Formula

**Confidence Interval = Sample Mean ± Margin of Error**

## Step 2 – Demonstrate the Concept

### Real-World Example

A company wants to estimate the average delivery time for all its customers.

Instead of collecting delivery times from every customer, it takes a sample.

Suppose the sample gives:

**Average = 30 minutes**

**95% Confidence Interval = 28 to 32 minutes**

The interval gives a range of plausible values for the actual population average.

### AI/ML Usage

Confidence intervals are useful for:

- Estimating population parameters
- Understanding uncertainty
- Comparing model performance
- Evaluating experimental results
- Statistical analysis

```python
import numpy as np
from scipy import stats

# Create sample delivery time data
delivery_times = np.array([28, 30, 31, 29, 32, 30, 27, 31, 29, 33])

# Calculate sample mean
sample_mean = np.mean(delivery_times)

# Calculate 95% confidence interval
confidence_interval = stats.t.interval(
    confidence=0.95,
    df=len(delivery_times) - 1,
    loc=sample_mean,
    scale=stats.sem(delivery_times)
)

print("Sample Mean:", sample_mean)
print("95% Confidence Interval:", confidence_interval)
```

## Step 3 – Explanation

### What does the program do?

The program calculates the sample mean and creates a **95% confidence interval** for the population mean.

### What statistical concept does it demonstrate?

It demonstrates how a confidence interval can be used to estimate a population parameter from sample data.

### How do we interpret the output?

The output gives a lower and upper limit.

The interval represents a range of plausible values for the population mean based on the sample.

### Where is this used in AI/ML?

Confidence intervals are used to:

- Measure uncertainty
- Estimate model metrics
- Compare experimental results
- Analyze sample data
- Support statistical decisions

### Important Point

A 95% confidence interval does not mean there is a 95% probability that the fixed population parameter is inside this particular interval. The 95% refers to the long-run performance of the interval procedure.

# Significance Level

## Step 1 – Understand the Concept

The **Significance Level (α)** is the threshold used to decide whether we have enough statistical evidence to reject the Null Hypothesis.

It is chosen **before performing the hypothesis test**.

### Common Significance Levels

- **0.10** → 10%
- **0.05** → 5%
- **0.01** → 1%

The most commonly used significance level is **0.05**.

### Decision Rule

- **p-value ≤ α** → Reject H₀
- **p-value > α** → Fail to reject H₀

### Simple Example

If:

**α = 0.05**

and

**p-value = 0.03**

Since:

**0.03 < 0.05**

we reject the Null Hypothesis.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company wants to check whether a new training program changes employee productivity.

Before testing, the company chooses:

**Significance Level (α) = 0.05**

After performing the test:

**p-value = 0.03**

Since the p-value is smaller than the significance level, there is enough statistical evidence to reject the Null Hypothesis.

### AI/ML Usage

Significance levels are used in:

- Hypothesis testing
- Feature analysis
- Comparing datasets
- Statistical experiments
- Evaluating model-related experiments

```python
# Set the significance level
alpha = 0.05

# Example p-value
p_value = 0.03

print("Significance Level:", alpha)
print("p-value:", p_value)

# Make a decision
if p_value <= alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

## Step 3 – Explanation

### What does the program do?

The program compares the p-value with the chosen significance level.

### What statistical concept does it demonstrate?

It demonstrates how the **Significance Level (α)** is used to make a decision in hypothesis testing.

### How do we interpret the output?

The significance level is **0.05** and the p-value is **0.03**.

Since:

**0.03 ≤ 0.05**

we reject the Null Hypothesis.

### Where is this used in AI/ML?

Significance levels are used when performing statistical tests to determine whether observed results provide enough evidence against the Null Hypothesis.

### Important Point

The significance level is a decision threshold. It does not measure the size or practical importance of an effect.

# Type I Error

## Step 1 – Understand the Concept

**Type I Error** occurs when we **reject the Null Hypothesis even though it is actually true**.

In simple words, we conclude that there is an effect or difference when there is actually no effect or difference.

### Also Called

Type I Error is also called a **False Positive**.

### Simple Example

Suppose a medical test checks whether a person has a disease.

- **H₀:** The person does not have the disease.
- The test says the person has the disease.
- But the person actually does not have the disease.

This is a **Type I Error**.

### Relationship with Significance Level

The probability of making a Type I Error is controlled by the **significance level (α)**.

For example:

**α = 0.05**

means the test is performed with a 5% significance threshold for Type I Error.

## Step 2 – Demonstrate the Concept

### Real-World Example

A fraud detection system checks whether a transaction is fraudulent.

- **H₀:** The transaction is not fraudulent.
- The system identifies it as fraudulent.
- But the transaction is actually genuine.

This is a **False Positive**, which is an example of Type I Error.

### AI/ML Usage

Type I Errors can occur in:

- Fraud detection
- Spam detection
- Medical diagnosis
- Anomaly detection
- Classification systems

### Key Idea

**Type I Error = False Positive**

We detect something as present when it is actually absent.

```python
# Set the significance level
alpha = 0.05

# Example p-value
p_value = 0.02

print("Significance Level:", alpha)
print("p-value:", p_value)

# Decision based on the p-value
if p_value <= alpha:
    print("Reject the Null Hypothesis")
    print("Possible Type I Error if H₀ is actually true.")
else:
    print("Fail to Reject the Null Hypothesis")
```

## Step 3 – Explanation

### What does the program do?

The program compares the p-value with the significance level and makes a hypothesis-testing decision.

### What statistical concept does it demonstrate?

It demonstrates the idea of a **Type I Error**.

### How do we interpret the output?

Since the p-value is less than 0.05, the program rejects the Null Hypothesis.

If the Null Hypothesis is actually true, this decision would be a **Type I Error**.

### Where is this used in AI/ML?

Type I Error is important when evaluating systems such as:

- Fraud detection
- Medical diagnosis
- Spam detection
- Anomaly detection

### Important Point

**Type I Error = Rejecting a true Null Hypothesis = False Positive**

# Type II Error

## Step 1 – Understand the Concept

**Type II Error** occurs when we **fail to reject the Null Hypothesis even though it is actually false**.

In simple words, we conclude that there is no effect or difference when there is actually an effect or difference.

### Also Called

Type II Error is also called a **False Negative**.

### Simple Example

Suppose a medical test checks whether a person has a disease.

- **H₀:** The person does not have the disease.
- The test says the person does not have the disease.
- But the person actually has the disease.

This is a **Type II Error**.

### Key Idea

**Type II Error = False Negative**

## Step 2 – Demonstrate the Concept

### Real-World Example

A fraud detection system checks whether a transaction is fraudulent.

- **H₀:** The transaction is not fraudulent.
- The system says the transaction is not fraudulent.
- But the transaction is actually fraudulent.

This is a **False Negative**, which is an example of Type II Error.

### AI/ML Usage

Type II Errors can occur in:

- Fraud detection
- Medical diagnosis
- Spam detection
- Anomaly detection
- Classification systems

### Key Difference

**Type I Error → False Positive**

**Type II Error → False Negative**

```python
# Example decision

actual_condition = True
test_result = False

print("Actual Condition:", actual_condition)
print("Test Result:", test_result)

# Check for Type II Error
if actual_condition and not test_result:
    print("Type II Error: False Negative")
else:
    print("No Type II Error")
```

## Step 3 – Explanation

### What does the program do?

The program compares the actual condition with the test result.

### What statistical concept does it demonstrate?

It demonstrates the concept of **Type II Error**.

### How do we interpret the output?

The actual condition is **True**, but the test result is **False**.

So, the system failed to identify an actual condition.

This is a **Type II Error (False Negative)**.

### Where is this used in AI/ML?

Type II Error is important in:

- Fraud detection
- Medical diagnosis
- Anomaly detection
- Spam detection
- Classification models

### Important Point

**Type II Error = Failing to reject a false Null Hypothesis = False Negative**

# One-Tailed Test

## Step 1 – Understand the Concept

A **One-Tailed Test** is a hypothesis test where we are interested in finding a difference in **one specific direction**.

The direction can be:

- **Greater than (>)**
- **Less than (<)**

### Example

A company claims that the average delivery time is **30 minutes**.

We want to test whether the new system **reduces** the delivery time.

**H₀:** Average delivery time ≥ 30 minutes

**H₁:** Average delivery time < 30 minutes

Here, we are only interested in the **lower direction**, so it is a one-tailed test.

### Key Point

A one-tailed test checks for an effect in **one direction only**.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company introduces a new delivery system and wants to know whether it **reduces** the average delivery time.

The hypotheses are:

**H₀:** The new system does not reduce delivery time.

**H₁:** The new system reduces delivery time.

Since we are specifically checking for a **reduction**, this is a one-tailed test.

### AI/ML Usage

One-tailed tests can be used when we have a specific directional question, such as:

- Whether a new model increases accuracy
- Whether a new method reduces error
- Whether a treatment decreases processing time
- Whether a feature increases a measured outcome

```python
from scipy import stats
import numpy as np

# Create sample delivery time data
delivery_times = np.array([27, 28, 29, 30, 28, 27, 29, 28, 30, 27])

# Test against an assumed mean of 30 minutes
t_statistic, p_value_two_tailed = stats.ttest_1samp(
    delivery_times, 30
)

# Convert to one-tailed p-value
if t_statistic < 0:
    p_value_one_tailed = p_value_two_tailed / 2
else:
    p_value_one_tailed = 1 - (p_value_two_tailed / 2)

print("Sample Mean:", np.mean(delivery_times))
print("T-Statistic:", t_statistic)
print("One-Tailed p-value:", p_value_one_tailed)

# Set significance level
alpha = 0.05

if p_value_one_tailed <= alpha and t_statistic < 0:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

## Step 3 – Explanation

### What does the program do?

The program checks whether the sample delivery time is significantly **less than 30 minutes**.

### What statistical concept does it demonstrate?

It demonstrates a **One-Tailed Test** where the alternative hypothesis focuses on one direction.

### How do we interpret the output?

We compare the one-tailed p-value with the significance level of **0.05**.

If the p-value is less than or equal to 0.05 and the result is in the expected direction, we reject the Null Hypothesis.

### Where is this used in AI/ML?

One-tailed tests can be used when we want to test a specific directional change, such as:

- Accuracy increases
- Error decreases
- Processing time decreases
- Performance improves

### Important Point

A one-tailed test should be chosen when the research question is **directional before the analysis is performed**.

# Two-Tailed Test

## Step 1 – Understand the Concept

A **Two-Tailed Test** is a hypothesis test where we are interested in finding a difference in **either direction**.

It checks whether the value is:

- Greater than the expected value
- Less than the expected value

### Example

A company claims that the average delivery time is **30 minutes**.

We want to know whether the actual delivery time is **different from 30 minutes**.

**H₀:** Average delivery time = 30 minutes

**H₁:** Average delivery time ≠ 30 minutes

Since we are checking both directions, this is a **Two-Tailed Test**.

### Key Point

A two-tailed test checks for a significant difference in **both directions**.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company changes its delivery system.

The company wants to know whether the new system changes the average delivery time.

The delivery time could:

- Increase
- Decrease

We are interested in both possibilities.

Therefore, we use a **Two-Tailed Test**.

### AI/ML Usage

Two-tailed tests can be used when checking whether:

- Model accuracy is different
- Prediction error is different
- A feature has a significant effect
- Two groups have different averages

### Difference

**One-Tailed Test → One direction**

**Two-Tailed Test → Both directions**

```python
from scipy import stats
import numpy as np

# Create sample delivery time data
delivery_times = np.array([28, 29, 31, 30, 32, 29, 31, 30, 28, 32])

# Test against an assumed mean of 30 minutes
t_statistic, p_value = stats.ttest_1samp(
    delivery_times, 30
)

print("Sample Mean:", np.mean(delivery_times))
print("T-Statistic:", t_statistic)
print("Two-Tailed p-value:", p_value)

# Set significance level
alpha = 0.05

# Make a decision
if p_value <= alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

## Step 3 – Explanation

### What does the program do?

The program compares the sample mean with the assumed population mean of 30 minutes using a two-tailed t-test.

### What statistical concept does it demonstrate?

It demonstrates a **Two-Tailed Test**.

### How do we interpret the output?

The p-value is compared with the significance level of **0.05**.

- **p-value ≤ 0.05** → Reject H₀
- **p-value > 0.05** → Fail to Reject H₀

### Where is this used in AI/ML?

Two-tailed tests can be used to determine whether there is a statistically significant difference without specifying whether the difference should be positive or negative.

### Important Point

A two-tailed test checks for differences in **both directions**.
