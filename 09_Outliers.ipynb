# What are Outliers?

## Step 1 – Understand the Concept

**Outliers** are data points that are significantly different from the other observations in a dataset.

They are values that appear unusually high or unusually low compared with the general pattern of the data.

### Simple Example

Suppose the ages of 6 people are:

**22, 24, 25, 23, 26, 85**

Here, **85** is much higher than the other values.

Therefore, 85 can be considered an **outlier**.

### Why are Outliers Important?

Outliers can:

- Affect the mean
- Increase variance and standard deviation
- Affect statistical analysis
- Influence some ML models
- Sometimes represent errors or unusual real-world events

### Important Point

An outlier is not always an error.

It may represent a genuine unusual observation.

## Step 2 – Demonstrate the Concept

### Real-World Example

Consider monthly spending of customers:

**₹1,000, ₹1,200, ₹1,500, ₹1,300, ₹1,100, ₹20,000**

Here, ₹20,000 is much higher than the other values.

It may be an outlier.

Possible reasons:

- Data entry error
- One-time expensive purchase
- High-value customer
- Unusual event

### AI/ML Example

Outliers can appear in:

- Customer transactions
- Sensor data
- Financial data
- Medical measurements
- Network traffic
- Machine performance data

Before building an ML model, we may need to identify and understand unusual observations.

```python
import numpy as np
import matplotlib.pyplot as plt

# Create customer spending data
spending = np.array([1000, 1200, 1500, 1300, 1100, 20000])

# Calculate basic statistics
mean_spending = np.mean(spending)
median_spending = np.median(spending)

print("Customer Spending:", spending)
print("Mean Spending:", mean_spending)
print("Median Spending:", median_spending)

# Visualize the data
plt.scatter(range(len(spending)), spending)

plt.xlabel("Customer")
plt.ylabel("Spending")
plt.title("Customer Spending with a Possible Outlier")

plt.show()
```

## Step 3 – Explanation

### What does the program do?

The program:

1. Creates customer spending data.
2. Calculates the mean.
3. Calculates the median.
4. Creates a scatter plot to visualize the values.

### What statistical concept does it demonstrate?

It demonstrates the concept of an **Outlier**.

The value ₹20,000 is much higher than the other spending values and may be an outlier.

### How do we interpret the output?

The mean is strongly affected by the unusually high value.

The median is less affected by the extreme value.

This shows why outliers are important when analyzing data.

### Where is this used in AI/ML?

Outlier detection is used in:

- Fraud detection
- Anomaly detection
- Sensor monitoring
- Financial analysis
- Data cleaning
- Machine learning preprocessing

### Important Point

We should not automatically remove every outlier.

First, we should understand **why the value is unusual** and whether it represents a real observation or an error.

# Types of Outliers

## Step 1 – Understand the Concept

Outliers can occur in different ways depending on the data and the pattern.

### 1. Univariate Outlier

An unusual value in **one variable**.

Example:

**10, 12, 11, 13, 100**

Here, 100 is an unusual value compared with the other values.

### 2. Multivariate Outlier

A data point may not look unusual when variables are considered separately, but the **combination of multiple variables** is unusual.

Example:

A person may have:

- Age = 25
- Income = ₹50,000

Both values may be normal individually.

But a particular combination of age and income may be unusual compared with the overall dataset.

### 3. Global Outlier

A data point is very different from most of the observations in the entire dataset.

### 4. Contextual Outlier

A value is unusual only in a particular **context or situation**.

Example:

A temperature of 35°C may be normal in summer but unusual for a particular winter location.

### 5. Collective Outlier

A group of observations is unusual when considered together, even though individual values may not look unusual.

## Step 2 – Demonstrate the Concept

### Real-World Examples

#### Univariate Outlier

A customer's transaction amount:

**₹500, ₹600, ₹550, ₹700, ₹50,000**

₹50,000 may be a univariate outlier.

#### Contextual Outlier

A temperature of 40°C may be normal in summer but unusual in winter.

#### Multivariate Outlier

A customer may have a normal:

- Purchase amount
- Purchase frequency

But the combination of both may be unusual compared with other customers.

### AI/ML Example

Different types of outliers can appear in:

- Fraud detection
- Network security
- Sensor monitoring
- Credit card transactions
- Customer behavior analysis
- Anomaly detection systems

Understanding the type of outlier helps us choose an appropriate detection method.

```python
import numpy as np
import matplotlib.pyplot as plt

# Create normal values and an extreme value
data = np.array([10, 12, 11, 13, 14, 12, 15, 100])

# Calculate the mean
mean_value = np.mean(data)

print("Data:", data)
print("Mean:", mean_value)

# Visualize the values
plt.scatter(range(len(data)), data)

plt.xlabel("Observation")
plt.ylabel("Value")
plt.title("Example of an Univariate Outlier")

plt.show()
```

### What statistical concept does it demonstrate?

It demonstrates a **Univariate Outlier**.

The value 100 is much larger than the other observations.

### How do we interpret the output?

Most values are between 10 and 15, while 100 is far away from the rest.

The scatter plot makes this unusual value easy to identify.

### Where is this used in AI/ML?

Understanding outlier types is useful for:

### Important Point

Different outlier types may require different detection techniques. Common techniques include **IQR, Z-Score, and visualization methods**.

# Detecting Outliers

## Step 1 – Understand the Concept

**Outlier Detection** is the process of identifying data points that are unusually different from the rest of the dataset.

### Common Methods for Detecting Outliers

- IQR Method
- Z-Score Method
- Box Plot
- Scatter Plot

# IQR Method

## Step 1 – Understand the Concept

**IQR (Interquartile Range)** is a statistical method used to measure the spread of the middle 50% of a dataset.

It is also commonly used to detect outliers.

### Formula

**IQR = Q3 - Q1**

Where:

- **Q1** = First Quartile (25th percentile)
- **Q3** = Third Quartile (75th percentile)

### Outlier Boundaries

**Lower Bound = Q1 - 1.5 × IQR**

**Upper Bound = Q3 + 1.5 × IQR**

Any value below the lower bound or above the upper bound is considered a possible outlier.

### Simple Example

Data:

10, 12, 11, 13, 14, 12, 15, 100

The IQR method can help identify **100** as an outlier.

## Step 2 – Demonstrate the Concept

Suppose a shop records the daily sales:

₹10,000, ₹12,000, ₹11,000, ₹13,000, ₹14,000, ₹12,000, ₹15,000, ₹1,00,000

Most sales are between ₹10,000 and ₹15,000.

The value ₹1,00,000 is much higher than the other values.

Using the IQR method, we calculate the normal range and check whether ₹1,00,000 falls outside that range.

### Real-World Uses

- Detect unusual sales values
- Find abnormal transaction amounts
- Identify unusual sensor readings
- Detect unusual customer spending

### AI/ML Usage

IQR is commonly used during **data preprocessing** to identify unusual values before training a Machine Learning model.

```python
import numpy as np

# Create a dataset
sales = np.array([10000, 12000, 11000, 13000, 14000, 12000, 15000, 100000])

# Calculate Q1 and Q3
Q1 = np.percentile(sales, 25)
Q3 = np.percentile(sales, 75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliers = sales[(sales < lower_bound) | (sales > upper_bound)]

print("Sales Data:", sales)
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

## Step 3 – Explanation

### What does the program do?

The program calculates Q1, Q3, and IQR and then creates lower and upper boundaries.

It checks whether any sales value falls outside these boundaries.

### What statistical concept does it demonstrate?

It demonstrates the **IQR Method for Outlier Detection**.

### How do we interpret the output?

The value **100000** is outside the upper boundary.

Therefore, it is identified as an **outlier**.

### Where is this used in AI/ML?

The IQR method can be used in:

- Data preprocessing
- Outlier detection
- Fraud detection
- Anomaly detection
- Sensor data analysis

### Important Point

An outlier should not be removed automatically.

We should first check whether it is a genuine value or a data error.

# Z-Score Method

## Step 1 – Understand the Concept

**Z-Score** tells us how far a data point is from the mean in terms of standard deviations.

### Formula

**Z = (X - Mean) / Standard Deviation**

Where:

- **X** = Data value
- **Mean** = Average of the data
- **Standard Deviation** = Measure of data spread

### Outlier Rule

A commonly used rule is:

- Z-Score > +3 → Possible outlier
- Z-Score < -3 → Possible outlier

### Simple Example

Data:

10, 12, 11, 13, 14, 12, 15, 100

The value **100** is far away from the other values, so it may have a high Z-Score.

## Step 2 – Demonstrate the Concept

Imagine a company checking employee salaries.

Most employees earn between ₹25,000 and ₹40,000.

If one employee earns ₹2,00,000, this value is much higher than the average.

The Z-Score helps us measure how far this value is from the average salary.

### Real-World Uses

- Detect unusual bank transactions
- Identify abnormal sensor readings
- Find unusual exam scores
- Detect unusual customer behavior

### AI/ML Usage

Z-Score is useful for:

- Outlier detection
- Data preprocessing
- Anomaly detection
- Feature scaling

```python
import numpy as np

# Create a dataset
data = np.array([10, 12, 11, 13, 14, 12, 15, 100])

# Calculate mean
mean = np.mean(data)

# Calculate standard deviation
std = np.std(data)

# Calculate Z-Scores
z_scores = (data - mean) / std

# Detect possible outliers
outliers = data[np.abs(z_scores) > 3]

print("Data:", data)
print("Mean:", mean)
print("Standard Deviation:", std)
print("Z-Scores:", z_scores)
print("Possible Outliers:", outliers)
```

## Step 3 – Explanation

### What does the program do?

The program calculates the mean and standard deviation and then calculates the Z-Score for each value.

### What statistical concept does it demonstrate?

It demonstrates **Z-Score based Outlier Detection**.

### How do we interpret the output?

The Z-Score tells us how far each value is from the mean.

A value with a Z-Score greater than +3 or less than -3 is commonly considered a possible outlier.

### Where is this used in AI/ML?

Z-Score is used in:

- Outlier detection
- Data preprocessing
- Anomaly detection
- Feature scaling
- Identifying unusual observations

# Box Plot

## Step 1 – Understand the Concept

A **Box Plot** is a graphical representation used to understand the distribution of data and identify possible outliers.

A box plot shows:

- **Minimum** – Smallest value
- **Q1** – First Quartile (25th percentile)
- **Median** – Middle value
- **Q3** – Third Quartile (75th percentile)
- **Maximum** – Largest value

### Outliers in a Box Plot

Values that appear separately from the main box and whiskers are usually considered **possible outliers**.

### Simple Example

If most customer spending values are between ₹1,000 and ₹5,000, but one customer spends ₹50,000, the box plot can show ₹50,000 as a possible outlier.

### Why Use Box Plots?

- Understand data distribution
- Compare datasets
- Identify possible outliers
- Understand spread and median

## Step 2 – Demonstrate the Concept

Suppose a company records the daily sales:

₹10,000, ₹12,000, ₹11,000, ₹13,000, ₹14,000, ₹12,000, ₹15,000, ₹1,00,000

Most sales are between ₹10,000 and ₹15,000.

The value ₹1,00,000 is much higher than the other values.

A box plot can visually show this value as a possible outlier.

### Real-World Uses

- Analyze sales data
- Compare employee salaries
- Identify unusual transaction values
- Analyze sensor measurements

### AI/ML Usage

Box plots are useful during **Exploratory Data Analysis (EDA)** to understand data distribution and identify possible outliers before building ML models.

```python
import numpy as np
import matplotlib.pyplot as plt

# Create a dataset
sales = np.array([10000, 12000, 11000, 13000, 14000, 12000, 15000, 100000])

# Create a box plot
plt.boxplot(sales)

plt.ylabel("Sales Amount")
plt.title("Box Plot of Daily Sales")

plt.show()
```

## Step 3 – Explanation

### What does the program do?

The program creates a box plot for the sales data.

### What statistical concept does it demonstrate?

It demonstrates how a **Box Plot** can be used to understand data distribution and identify possible outliers.

### How do we interpret the output?

The main box represents the middle part of the data.

The value **100000** appears separately from the main distribution, indicating that it is a possible outlier.

### Where is this used in AI/ML?

Box plots are used in:

- Exploratory Data Analysis
- Outlier detection
- Data preprocessing
- Comparing different datasets
- Understanding feature distributions

# Scatter Plot

## Step 1 – Understand the Concept

A **Scatter Plot** is a graph used to show the relationship between two numerical variables.

Each point on the graph represents one observation.

Scatter plots can also help us identify **possible outliers**.

### Simple Example

Suppose we have:

- X-axis → Study Hours
- Y-axis → Exam Score

Most students may have study hours between 2 and 6 hours.

If one student has 20 study hours but a very different score, that point may be an unusual observation.

### Why Use Scatter Plots?

- Understand relationships between variables
- Identify patterns
- Find possible outliers
- Visualize data

## Step 2 – Demonstrate the Concept

Suppose we are analyzing customer data.

- X-axis → Number of Purchases
- Y-axis → Total Spending

Most customers may have a normal relationship between purchases and spending.

If one customer has a very high spending value compared with the others, the point may appear far away from the main group.

### Real-World Uses

- Customer spending analysis
- Sales analysis
- Sensor data analysis
- Financial data analysis

### AI/ML Usage

Scatter plots are useful in:

- Exploratory Data Analysis (EDA)
- Outlier detection
- Feature relationship analysis
- Identifying unusual observations

```python
import numpy as np
import matplotlib.pyplot as plt

# Create number of purchases
purchases = np.array([2, 3, 4, 5, 6, 7, 8, 20])

# Create total spending
spending = np.array([1000, 1500, 1800, 2200, 2500, 2800, 3000, 15000])

# Create a scatter plot
plt.scatter(purchases, spending)

plt.xlabel("Number of Purchases")
plt.ylabel("Total Spending")
plt.title("Customer Purchases vs Spending")

plt.show()
```

## Step 3 – Explanation

### What does the program do?

The program creates a scatter plot showing the relationship between the number of purchases and total spending.

### What statistical concept does it demonstrate?

It demonstrates how a **Scatter Plot** can be used to visualize relationships and identify possible outliers.

### How do we interpret the output?

Most points are grouped together.

The point with **20 purchases and ₹15,000 spending** is far away from the other points, so it may be a possible outlier.

### Where is this used in AI/ML?

Scatter plots are used in:

- Exploratory Data Analysis
- Outlier detection
- Feature analysis
- Finding relationships between variables
- Understanding datasets before ML model training
