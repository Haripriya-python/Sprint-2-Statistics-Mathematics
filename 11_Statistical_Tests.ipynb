# Topic 1 – T-Test

## Definition

A T-Test is a statistical test used to determine whether there is a significant difference between means.

It is commonly used when the population standard deviation is unknown.

### Types of T-Test

1. **One-Sample T-Test**
   - Compares the mean of one sample with a known or expected value.

2. **Independent Two-Sample T-Test**
   - Compares the means of two independent groups.

3. **Paired T-Test**
   - Compares two related measurements, such as before and after values.

### Assumptions

- The data should be numerical.
- Observations should be independent when required.
- The data should be approximately normally distributed.
- For an independent T-Test, the groups should generally have similar variances when using the standard version.

### When to Use

- To compare means.
- When the population standard deviation is unknown.
- When comparing one sample, two independent groups, or paired measurements.

### When Not to Use

- When the data is categorical.
- When the required independence assumption is not satisfied.
- When the data is strongly non-normal, especially with a very small sample.
- When comparing more than two independent groups. ANOVA may be more appropriate.

### Business Example

A company wants to know whether employee productivity changed after a training program.

The company records productivity scores for the same employees before and after training.

A **Paired T-Test** can be used to determine whether the average productivity changed significantly after the training.

```python
# Import required libraries
import numpy as np
from scipy.stats import ttest_rel

# Employee productivity scores before training
before = np.array([70, 72, 68, 75, 71, 69, 73, 70])

# Employee productivity scores after training
after = np.array([75, 76, 72, 78, 74, 73, 77, 75])

# Perform paired T-Test
t_statistic, p_value = ttest_rel(before, after)

# Display the results
print("T-Statistic:", t_statistic)
print("P-Value:", p_value)
```

## Code Explanation

### What the Program Does

The program compares employee productivity scores before and after training.

Since the same employees are measured before and after training, a paired T-Test is used.

### Statistical Concept Demonstrated

The program demonstrates a paired T-Test.

It checks whether there is a statistically significant difference between the average productivity scores before and after training.

### Python Functions Used

- `np.array()` creates numerical arrays containing the productivity scores.
- `ttest_rel()` performs a paired T-Test.
- `t_statistic` represents the calculated T-statistic.
- `p_value` represents the probability used to assess statistical significance.

### Interpretation

We use a significance level of 0.05.

- If p-value < 0.05 → Reject the null hypothesis. There is statistically significant evidence of a difference in productivity.
- If p-value >= 0.05 → Do not reject the null hypothesis. There is not enough statistical evidence of a difference.

The T-statistic shows the difference between the paired measurements relative to the variation in the data.

The p-value should be checked from the Python output to determine the conclusion for this example.

# Topic 2 – Z-Test

## Definition

A Z-Test is a statistical test used to determine whether there is a significant difference between a sample mean and a population mean, or between two population means.

It is commonly used when the population standard deviation is known or when the sample size is sufficiently large.

### Types of Z-Test

1. **One-Sample Z-Test**
   - Compares the mean of one sample with a known population mean.

2. **Two-Sample Z-Test**
   - Compares the means of two independent groups.

### Assumptions

- The data should be numerical.
- Observations should be independent.
- The population standard deviation should be known for the standard one-sample Z-Test.
- The data should be approximately normally distributed, especially for smaller samples.
- For large samples, the Central Limit Theorem can support the use of a Z-Test.

### When to Use

- To compare a sample mean with a known population mean.
- To compare the means of two independent groups.
- When the population standard deviation is known.
- When the sample size is large.

### When Not to Use

- When the population standard deviation is unknown and the sample size is small. A T-Test may be more appropriate.
- When the data is categorical.
- When observations are not independent.
- When the assumptions required for the Z-Test are not satisfied.

### Business Example

A company claims that the average delivery time is 30 minutes.

A sample of deliveries is collected, and the company wants to determine whether the actual average delivery time is significantly different from 30 minutes.

A One-Sample Z-Test can be used when the population standard deviation is known.

### Python Example

```python
# Import required libraries
import numpy as np
from scipy.stats import norm

# Sample delivery times
delivery_times = np.array([28, 32, 31, 29, 35, 30, 33, 27, 31, 34])

# Known population mean
population_mean = 30

# Known population standard deviation
population_std = 4

# Calculate sample mean
sample_mean = np.mean(delivery_times)

# Calculate sample size
n = len(delivery_times)

# Calculate Z-statistic
z_statistic = (sample_mean - population_mean) / (population_std / np.sqrt(n))

# Calculate two-tailed p-value
p_value = 2 * (1 - norm.cdf(abs(z_statistic)))

# Display the results
print("Sample Mean:", sample_mean)
print("Z-Statistic:", z_statistic)
print("P-Value:", p_value)
```

## Code Explanation

### What the Program Does

The program analyzes delivery times and compares the sample average delivery time with the claimed population average of 30 minutes.

It calculates the Z-statistic and p-value to determine whether the difference is statistically significant.

### Statistical Concept Demonstrated

The program demonstrates a One-Sample Z-Test.

The Z-Test measures how far the sample mean is from the population mean in terms of standard errors.

### Python Functions Used

- `np.array()` creates the delivery-time dataset.
- `np.mean()` calculates the sample mean.
- `len()` calculates the number of observations.
- `np.sqrt()` calculates the square root needed in the Z-statistic formula.
- `norm.cdf()` calculates the cumulative probability from the standard normal distribution.

### Interpretation

We use a significance level of 0.05.

- If p-value < 0.05 → Reject the null hypothesis. There is statistically significant evidence that the average delivery time is different from 30 minutes.
- If p-value >= 0.05 → Do not reject the null hypothesis. There is not enough statistical evidence that the average delivery time is different from 30 minutes.

The Z-statistic shows how many standard errors the sample mean is away from the population mean.

The p-value from the Python output should be used to make the final statistical conclusion.

# Topic 3 – Chi-Square Test

## Definition

The Chi-Square Test is a statistical test used to determine whether there is a significant relationship between categorical variables or whether observed frequencies differ from expected frequencies.

### Types of Chi-Square Test

1. **Chi-Square Test of Independence**
   - Determines whether two categorical variables are related.

2. **Chi-Square Goodness-of-Fit Test**
   - Determines whether observed frequencies match expected frequencies.

### Assumptions

- The data should be categorical.
- Observations should be independent.
- The categories should be mutually exclusive.
- Expected frequencies should generally be sufficiently large for the standard Chi-Square approximation.

### When to Use

- To test the relationship between two categorical variables.
- To compare observed and expected frequencies.
- When the data is represented as counts or frequencies.

### When Not to Use

- When the data is continuous without being converted into appropriate categories.
- When observations are not independent.
- When expected frequencies are too small for the Chi-Square approximation.
- When you want to compare numerical means.

### Business Example

A company wants to know whether customer preference for a product is related to age group.

The company collects customer data and records age groups and product preferences.

A Chi-Square Test of Independence can be used to determine whether there is a statistically significant relationship between age group and product preference.

```python
# Import required library
from scipy.stats import chi2_contingency

# Customer preference data
# Rows represent age groups
# Columns represent Product A and Product B
data = [
    [30, 20],
    [25, 25],
    [15, 35]
]

# Perform Chi-Square Test of Independence
chi2_statistic, p_value, degrees_of_freedom, expected = chi2_contingency(data)

# Display the results
print("Chi-Square Statistic:", chi2_statistic)
print("P-Value:", p_value)
print("Degrees of Freedom:", degrees_of_freedom)

print("\nExpected Frequencies:")
print(expected)
```

## Code Explanation

### What the Program Does

The program analyzes customer age groups and their product preferences.

It uses the observed customer counts to determine whether age group and product preference are statistically related.

### Statistical Concept Demonstrated

The program demonstrates the **Chi-Square Test of Independence**.

It compares the observed frequencies with the expected frequencies that would occur if the two categorical variables were independent.

### Python Functions Used

- `chi2_contingency()` performs the Chi-Square Test of Independence.
- `chi2_statistic` represents the calculated Chi-Square value.
- `p_value` is used to determine statistical significance.
- `degrees_of_freedom` represents the degrees of freedom.
- `expected` contains the expected frequencies.

### Interpretation

We use a significance level of 0.05.

- If p-value < 0.05 → Reject the null hypothesis. There is statistically significant evidence of a relationship between age group and product preference.
- If p-value >= 0.05 → Do not reject the null hypothesis. There is not enough statistical evidence of a relationship.

The Chi-Square statistic measures the difference between the observed and expected frequencies.

The p-value from the Python output should be used to make the final conclusion.

# Topic 4 – ANOVA

## Definition

ANOVA stands for Analysis of Variance.

It is a statistical test used to determine whether there is a significant difference between the means of three or more independent groups.

ANOVA compares the variation between groups with the variation within groups.

### Types of ANOVA

1. **One-Way ANOVA**
   - Compares the means of three or more independent groups using one factor.

2. **Two-Way ANOVA**
   - Examines the effect of two factors on a numerical outcome.

### Assumptions

- The dependent variable should be numerical.
- The groups should be independent.
- The data in each group should be approximately normally distributed.
- The groups should have similar variances.
- Observations should be independent.

### When to Use

- When comparing the means of three or more groups.
- When the groups are independent.
- When the dependent variable is numerical.

### When Not to Use

- When there are only two groups; a T-Test may be more suitable.
- When the data is categorical.
- When observations are not independent.
- When the assumptions of ANOVA are not reasonably satisfied.

### Business Example

A company wants to compare the average sales generated by employees using three different sales training programs.

The company records the sales performance of employees from each training group.

One-Way ANOVA can be used to determine whether there is a statistically significant difference in the average sales among the three training groups.

```python
# Import required library
from scipy.stats import f_oneway

# Sales performance for three training groups
training_a = [70, 72, 75, 68, 71]
training_b = [78, 80, 82, 79, 81]
training_c = [65, 67, 64, 68, 66]

# Perform One-Way ANOVA
f_statistic, p_value = f_oneway(
    training_a,
    training_b,
    training_c
)

# Display the results
print("F-Statistic:", f_statistic)
print("P-Value:", p_value)
```

## Code Explanation

### What the Program Does

The program compares the sales performance of employees from three different training groups.

It uses One-Way ANOVA to determine whether the average sales performance is significantly different among the groups.

### Statistical Concept Demonstrated

The program demonstrates **One-Way ANOVA**.

ANOVA compares the variation between the group means with the variation within the groups.

### Python Functions Used

- `f_oneway()` performs the One-Way ANOVA test.
- `f_statistic` represents the F-statistic.
- `p_value` is used to determine whether the difference between group means is statistically significant.

### Interpretation

We use a significance level of 0.05.

- If p-value < 0.05 → Reject the null hypothesis. There is statistically significant evidence that at least one group mean is different.
- If p-value >= 0.05 → Do not reject the null hypothesis. There is not enough statistical evidence of a difference between the group means.

The F-statistic compares the variation between groups with the variation within groups.

If ANOVA gives a significant result, additional tests such as a post-hoc test can be used to identify which specific groups are different.

### AI/ML Usage

ANOVA can be used during data analysis to compare the effect of different groups or categories on a numerical outcome.

# Topic 5 – Mann-Whitney U Test

## Definition

The Mann-Whitney U Test is a non-parametric statistical test used to compare two independent groups.

It checks whether the distributions or ranks of the two groups are significantly different.

It is an alternative to the Independent Two-Sample T-Test when the assumptions of the T-Test are not satisfied.

### Assumptions

- The two groups should be independent.
- The data should be at least ordinal.
- The observations should be independent within each group.
- The two groups should be measured on a comparable scale.

### When to Use

- When comparing two independent groups.
- When the data is not normally distributed.
- When the sample size is small.
- When the data is ordinal or numerical but does not satisfy the assumptions of a T-Test.

### When Not to Use

- When the two groups are paired or related.
- When there are more than two independent groups.
- When observations are not independent.
- When the data does not provide meaningful ranking or ordering.

### Business Example

A company wants to compare customer satisfaction ratings for two different products.

The ratings are collected from two independent groups of customers.

A Mann-Whitney U Test can be used to determine whether there is a statistically significant difference in the customer satisfaction ratings between the two products.

```python
# Import required library
from scipy.stats import mannwhitneyu

# Customer satisfaction ratings for two products
product_a = [7, 8, 6, 9, 7, 8, 6]
product_b = [5, 6, 5, 7, 6, 5, 6]

# Perform Mann-Whitney U Test
u_statistic, p_value = mannwhitneyu(
    product_a,
    product_b,
    alternative="two-sided"
)

# Display the results
print("U-Statistic:", u_statistic)
print("P-Value:", p_value)
```

## Code Explanation

### What the Program Does

The program compares customer satisfaction ratings for two independent products.

It uses the Mann-Whitney U Test to determine whether the ratings of the two products are statistically different.

### Statistical Concept Demonstrated

The program demonstrates the Mann-Whitney U Test.

The test ranks the observations and compares the ranks between the two independent groups.

### Python Functions Used

- `mannwhitneyu()` performs the Mann-Whitney U Test.
- `u_statistic` represents the calculated U-statistic.
- `p_value` is used to determine statistical significance.
- `alternative="two-sided"` checks for a difference in either direction.

### Interpretation

We use a significance level of 0.05.

- If p-value < 0.05 → Reject the null hypothesis. There is statistically significant evidence of a difference between the two groups.
- If p-value >= 0.05 → Do not reject the null hypothesis. There is not enough statistical evidence of a difference between the two groups.

The U-statistic represents the comparison of the ranks between the two groups.

### AI/ML Usage

The Mann-Whitney U Test can be used during data analysis when comparing two independent groups and the numerical data does not satisfy the assumptions required for a parametric test such as the T-Test.

# Topic 6 – Shapiro-Wilk Test

## Definition

The Shapiro-Wilk Test is a statistical test used to determine whether a dataset is approximately normally distributed.

It is commonly used before applying statistical tests that assume normality, such as the T-Test and ANOVA.

### Assumptions

- The data should be numerical.
- Observations should be independent.
- The test is mainly used for checking whether the data follows a normal distribution.
- The sample should be suitable for the Shapiro-Wilk test.

### When to Use

- To check whether numerical data is approximately normally distributed.
- Before using statistical tests that require a normality assumption.
- During data analysis and preprocessing.

### When Not to Use

- When the data is categorical.
- When observations are not independent.
- When normality is not relevant to the analysis.
- For very large datasets, where even small departures from normality may become statistically significant.

### Business Example

A company wants to compare employee productivity scores using a T-Test.

Before performing the T-Test, the company checks whether the productivity scores are approximately normally distributed.

A Shapiro-Wilk Test can be used to check the normality of the productivity data.

```python
# Import required library
from scipy.stats import shapiro

# Employee productivity scores
productivity = [72, 75, 71, 73, 76, 74, 70, 77, 73, 75]

# Perform Shapiro-Wilk Test
statistic, p_value = shapiro(productivity)

# Display the results
print("Shapiro-Wilk Statistic:", statistic)
print("P-Value:", p_value)
```

## Code Explanation

### What the Program Does

The program checks whether the employee productivity scores are approximately normally distributed.

It uses the Shapiro-Wilk Test and calculates the test statistic and p-value.

### Statistical Concept Demonstrated

The program demonstrates the Shapiro-Wilk Test.

The test checks whether the observed data is consistent with a normal distribution.

### Python Functions Used

- `shapiro()` performs the Shapiro-Wilk normality test.
- `statistic` represents the Shapiro-Wilk test statistic.
- `p_value` is used to determine whether there is evidence against normality.

### Interpretation

We use a significance level of 0.05.

- If p-value < 0.05 → Reject the null hypothesis. There is statistically significant evidence that the data is not normally distributed.
- If p-value >= 0.05 → Do not reject the null hypothesis. There is not enough statistical evidence to conclude that the data is not normally distributed.

The Shapiro-Wilk statistic indicates how closely the data follows a normal distribution.

### AI/ML Usage

The Shapiro-Wilk Test can be used during exploratory data analysis to check the normality of numerical features before applying statistical methods that assume normally distributed data.
