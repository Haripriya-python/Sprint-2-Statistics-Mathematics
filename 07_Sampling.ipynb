# Sampling

## Step 1 – Understand the Concept

**Sampling** is the process of selecting a smaller group of people or observations from a larger population to study and understand the population.

### Important Terms

- **Population** – The complete group we want to study.
- **Sample** – A smaller group selected from the population.
- **Sampling** – The process of selecting the sample.

### Simple Example

Suppose a company has **10,000 customers** and wants to know their satisfaction level.

Instead of asking all 10,000 customers, the company may select **500 customers** and collect their feedback.

Here:

- Population = 10,000 customers
- Sample = 500 customers
- Sampling = Process of selecting those 500 customers

The goal is to use information from the sample to understand the larger population.

## Step 2 – Demonstrate the Concept

Imagine a school has **1,000 students**.

The school wants to know the average number of hours students spend studying every day.

Instead of asking all 1,000 students, the school selects **100 students** and collects their study hours.

### Population

`1,000 students`

### Sample

`100 selected students`

### Purpose of Sampling

The sample is used to estimate or understand characteristics of the entire population.

### Real-World Examples

Sampling is used in:

- Customer surveys
- Election surveys
- Market research
- Medical studies
- Quality testing
- Machine learning datasets

In AI/ML, working with a representative sample can reduce the amount of data that needs to be processed while still providing useful information about the larger dataset.

```python
import numpy as np

# Create a population of 1,000 students
population = np.arange(1, 1001)

# Select a sample of 100 students randomly
sample = np.random.choice(population, size=100, replace=False)

# Display the sample size
print("Population size:", len(population))
print("Sample size:", len(sample))

# Display the first 10 selected students
print("First 10 sampled students:", sample[:10])
```

## Step 3 – Explanation

### What does the program do?

The program creates a population containing 1,000 students.

It then randomly selects 100 students from the population without selecting the same student twice.

### What statistical concept does it demonstrate?

It demonstrates **Sampling**.

Instead of working with the entire population, we select a smaller group called a sample.

### How do we interpret the output?

The output will show:

- Population size = 1,000
- Sample size = 100

The selected student IDs will be different each time because the sample is selected randomly.

### Where is this used in AI/ML?

Sampling is useful in AI/ML for:

- Working with large datasets
- Exploratory Data Analysis (EDA)
- Model development
- Testing and validation
- Reducing computation time
- Creating representative datasets

For very large datasets, using an appropriate sample can make data analysis faster and easier.

# Random Sampling

## Step 1 – Understand the Concept

**Random Sampling** is a sampling method where every member of the population has an equal or known chance of being selected.

The selection is done randomly, so the researcher does not intentionally choose specific individuals.

### Simple Example

Suppose a company has **1,000 employees** and wants to survey 100 employees.

Using random sampling, the company randomly selects 100 employees from the 1,000.

Each employee has the same chance of being selected.

### Key Point

Random sampling helps reduce **selection bias** because the researcher does not manually choose who should be included in the sample.

## Step 2 – Demonstrate the Concept

Suppose a college has 1,000 students.

The college wants to collect feedback from 100 students.

Instead of selecting students based on:

- Name
- Department
- Marks
- Personal preference

we randomly select 100 students.

### Population

`1,000 students`

### Sample

`100 randomly selected students`

### Real-World Examples

Random sampling is used in:

- Customer surveys
- Market research
- Quality testing
- Social research
- Medical studies
- Dataset analysis

### AI/ML Example

Suppose a dataset contains **1 million customer records**.

We can randomly select **100,000 records** for initial analysis or experimentation.

This can reduce computation while giving a sample of the larger dataset.

```python
import numpy as np

# Create a population of 1,000 customers
population = np.arange(1, 1001)

# Select 100 customers randomly
random_sample = np.random.choice(
    population,
    size=100,
    replace=False
)

# Display the sample
print("Population size:", len(population))
print("Sample size:", len(random_sample))

# Display the first 10 selected customers
print("First 10 selected customers:", random_sample[:10])
```

## Step 3 – Explanation

### What does the program do?

The program creates a population of 1,000 customers and randomly selects 100 customers from that population.

`replace=False` ensures that the same customer is not selected more than once.

### What statistical concept does it demonstrate?

It demonstrates **Random Sampling**, where members are selected randomly from the population.

### How do we interpret the output?

The output will show:

- Population size = 1,000
- Sample size = 100
- 100 randomly selected customer IDs

The selected IDs may be different each time the program runs.

### Where is this used in AI/ML?

Random sampling can be used for:

- Creating training and testing datasets
- Exploratory Data Analysis (EDA)
- Working with large datasets
- Experimentation
- Data preprocessing
- Reducing computational workload

It is useful when we want a sample that represents the larger population without manually selecting specific records.

# Stratified Sampling

## Step 1 – Understand the Concept

**Stratified Sampling** is a sampling method where the population is divided into smaller groups called **strata** based on a specific characteristic.

Then, samples are selected from **each group**.

### Simple Example

Suppose a college has 1,000 students:

- 600 Undergraduate students
- 400 Postgraduate students

If we want a sample of 100 students, we can select students from both groups.

The groups are called **strata**.

### Key Point

Stratified sampling ensures that important groups in the population are represented in the sample.

Common characteristics used to create strata include:

- Age
- Gender
- Department
- Education level
- Location
- Income group

## Step 2 – Demonstrate the Concept

Suppose a company has 1,000 employees:

| Department | Employees |
|------------|-----------|
| IT         | 500       |
| HR         | 200       |
| Finance    | 150       |
| Marketing  | 150       |

The company wants to select 100 employees.

Instead of selecting 100 employees from the entire population randomly, we can select employees from **each department**.

For example:

- IT → 50 employees
- HR → 20 employees
- Finance → 15 employees
- Marketing → 15 employees

This keeps the sample proportionally similar to the population.

### AI/ML Example

Suppose a dataset contains customers from different age groups.

We can divide customers into:

- Young
- Middle-aged
- Senior

Then select samples from each group.

This helps ensure that important groups are represented in the dataset.

```python
import pandas as pd

# Create employee data
employees = pd.DataFrame({
    "Employee_ID": range(1, 11),
    "Department": [
        "IT", "IT", "IT",
        "HR", "HR",
        "Finance", "Finance",
        "Marketing", "Marketing", "Marketing"
    ]
})

# Select a sample from each department
stratified_sample = (
    employees
    .groupby("Department", group_keys=False)
    .apply(lambda group: group.sample(n=1, random_state=42))
)

# Display the stratified sample
print("Stratified Sample:")
print(stratified_sample)
```

## Step 3 – Explanation

### What does the program do?

The program creates an employee dataset containing different departments.

It divides the employees into groups based on their department and randomly selects one employee from each group.

### What statistical concept does it demonstrate?

It demonstrates **Stratified Sampling**.

The population is divided into groups, called strata, and samples are selected from each group.

### How do we interpret the output?

The output contains employees from:

- IT
- HR
- Finance
- Marketing

Each department is represented in the sample.

### Where is this used in AI/ML?

Stratified sampling is useful when a dataset contains important groups that should be represented.

It is commonly used for:

- Classification datasets
- Train-test splitting
- Customer analysis
- Survey data
- Imbalanced datasets
- Model evaluation

For example, when creating a train-test split for a classification problem, stratified sampling can help preserve the proportion of different classes in both datasets.

# Cluster Sampling

## Step 1 – Understand the Concept

**Cluster Sampling** is a sampling method where the population is divided into natural groups called **clusters**.

Instead of selecting individual people from the entire population, we randomly select some clusters and study the members within those selected clusters.

### Simple Example

A city has 20 schools.

- Population → All students in 20 schools
- Clusters → Each school
- Select 5 schools randomly
- Collect data from students in those selected schools

### Stratified vs Cluster Sampling

- **Stratified Sampling:** Select samples from every group.
- **Cluster Sampling:** Select some groups and study the selected groups.

### Real-World Examples

- Selecting schools for an education survey
- Selecting branches of a bank for customer analysis
- Selecting villages for a population survey
- Selecting geographic regions for market research

## Step 2 – Demonstrate the Concept

### Example

Suppose a company has customers from 4 branches:

- Chennai
- Salem
- Madurai
- Coimbatore

Each branch is treated as a **cluster**.

We randomly select 2 branches and include the customers from those selected branches in our sample.

### AI/ML Example

Cluster sampling can be useful when data is naturally grouped by:

- City
- Branch
- School
- Hospital
- Region

It can reduce the cost and effort of collecting data from a very large population.

```python
import pandas as pd

# Create customer data
customers = pd.DataFrame({
    "Customer_ID": range(1, 13),
    "Branch": [
        "Chennai", "Chennai", "Chennai",
        "Salem", "Salem", "Salem",
        "Madurai", "Madurai", "Madurai",
        "Coimbatore", "Coimbatore", "Coimbatore"
    ]
})

# Randomly select 2 branches as clusters
selected_clusters = customers["Branch"].drop_duplicates().sample(
    n=2,
    random_state=42
)

# Select all customers from the chosen clusters
cluster_sample = customers[
    customers["Branch"].isin(selected_clusters)
]

print("Selected Clusters:")
print(selected_clusters.tolist())

print("\nCluster Sample:")
print(cluster_sample)
```

## Step 3 – Explanation

### What does the program do?

The program:

1. Creates customer data for 4 branches.
2. Treats each branch as a cluster.
3. Randomly selects 2 branches.
4. Selects all customers belonging to those branches.

### What statistical concept does it demonstrate?

It demonstrates **Cluster Sampling**, where some natural groups are randomly selected instead of selecting individual members from the entire population.

### How do we interpret the output?

The output shows:

- Which branches were selected as clusters.
- All customers belonging to those selected branches.

The selected clusters can change if the random selection changes.

### Where is this used in AI/ML?

Cluster sampling can be used for:

- Large-scale dataset collection
- Geographic data analysis
- Customer data analysis
- Survey data
- Data collection from different branches or regions

It can reduce data collection cost and effort, but selecting only a few clusters may increase sampling error if those clusters are not representative of the whole population.

# Systematic Sampling

## Step 1 – Understand the Concept

**Systematic Sampling** is a sampling method where we select data members at a regular interval from a population.

Instead of randomly selecting every individual, we choose a starting point and then select every **k-th** member.

### Formula

Sampling Interval:

**k = Population Size / Sample Size**

### Simple Example

Suppose we have 1,000 customers and need a sample of 100 customers.

k = 1000 / 100 = 10

We can select:

10, 20, 30, 40, 50, ...

So, every 10th customer is selected.

### Real-World Examples

- Selecting every 10th customer from a customer list
- Checking every 20th product in a manufacturing process
- Selecting every 5th record from a large dataset
- Surveying every 10th household

## Step 2 – Demonstrate the Concept

### Example

A company has 100 customers and wants to select 10 customers.

Sampling interval:

k = 100 / 10 = 10

We can select every 10th customer.

### AI/ML Example

Systematic sampling can be useful when a dataset is:

- Very large
- Ordered in a meaningful way
- Stored as a sequence of records

It can provide a simple and efficient way to select a sample without checking every record.

```python
import numpy as np

# Create a population of 100 customers
population = np.arange(1, 101)

# Define the required sample size
sample_size = 10

# Calculate the sampling interval
interval = len(population) // sample_size

# Select every 10th customer
systematic_sample = population[interval - 1::interval]

print("Population size:", len(population))
print("Sample size:", len(systematic_sample))
print("Sampling interval:", interval)
print("Systematic Sample:", systematic_sample)
```

## Step 3 – Explanation

### What does the program do?

The program:

1. Creates 100 customers.
2. Defines a sample size of 10.
3. Calculates the sampling interval.
4. Selects every 10th customer.

### What statistical concept does it demonstrate?

It demonstrates **Systematic Sampling**, where members are selected at a fixed interval.

### How do we interpret the output?

The output contains customers selected at regular intervals.

For example:

10, 20, 30, 40, 50, ...

This gives us a smaller sample from the larger population.

### Where is this used in AI/ML?

Systematic sampling can be used for:

- Large dataset sampling
- Data preprocessing
- Quality checking
- Customer data analysis
- Selecting records for model development

It is simple and efficient, especially when the data is already arranged in a sequence.

# Sampling Bias

## Step 1 – Understand the Concept

**Sampling Bias** occurs when the selected sample does not properly represent the entire population.

This happens when some members of the population have a higher or lower chance of being selected.

### Simple Example

Suppose a company wants to know how satisfied **all customers** are.

If the company collects feedback only from customers who purchased recently, the sample may not represent all customers.

### Common Causes

- Selecting only one type of person
- Excluding certain groups
- Using an unbalanced sampling method
- Collecting data from an easily available group

### Why is it important?

A biased sample can produce misleading results and can affect statistical analysis and AI/ML models.

## Step 2 – Demonstrate the Concept

### Example

Suppose a college has:

- 80 students from different departments
- 20 students from Computer Science
- 60 students from other departments

If we select only Computer Science students for a survey about the entire college, the sample is biased.

The sample does not represent the complete student population.

### AI/ML Example

Sampling bias can affect an AI/ML model when the training data does not represent the real-world users.

For example:

A face recognition model trained mostly using images from one demographic group may perform differently on groups that were underrepresented in the training data.

```python
import pandas as pd

# Create student population
students = pd.DataFrame({
    "Student_ID": range(1, 11),
    "Department": [
        "Computer Science", "Computer Science",
        "Computer Science", "Computer Science",
        "Computer Science",
        "Commerce", "Commerce",
        "Physics", "Physics", "Physics"
    ]
})

# Select only Computer Science students
biased_sample = students[
    students["Department"] == "Computer Science"
]

print("Total Population:")
print(students)

print("\nBiased Sample:")
print(biased_sample)

print("\nPopulation Size:", len(students))
print("Sample Size:", len(biased_sample))
```

## Step 3 – Explanation

### What does the program do?

The program creates a student population containing multiple departments.

It then selects only Computer Science students as the sample.

### What statistical concept does it demonstrate?

It demonstrates **Sampling Bias** because the sample represents only one department instead of the complete student population.

### How do we interpret the output?

The population contains students from:

- Computer Science
- Commerce
- Physics

But the sample contains only Computer Science students.

Therefore, conclusions based on this sample may not represent the entire college.

### Where is this used in AI/ML?

Sampling bias is important when:

- Creating training datasets
- Collecting customer data
- Building recommendation systems
- Developing classification models
- Evaluating model performance

If the training data is biased, the AI/ML model may also produce biased or less representative results.

# Sampling Error

## Step 1 – Understand the Concept

**Sampling Error** is the difference between the result obtained from a sample and the actual result of the entire population.

Sampling error happens because we study only a **sample**, not the complete population.

### Simple Example

Suppose a company has 10,000 customers.

The actual average customer spending is ₹2,000.

If we select a sample of 500 customers, the sample average may be ₹1,950.

The difference:

**Sampling Error = Sample Result − Population Result**

**Sampling Error = ₹1,950 − ₹2,000 = −₹50**

### Important Point

Sampling error is different from sampling bias.

- **Sampling Error:** Natural difference caused by using a sample.
- **Sampling Bias:** Systematic problem caused by selecting a sample that does not properly represent the population.

## Step 2 – Demonstrate the Concept

### Example

Suppose we have the following population of customer spending:

100, 200, 300, 400, 500

The population average can be calculated.

Now, we select only a few customers as our sample.

The sample average may be different from the population average.

This difference is called **Sampling Error**.

### AI/ML Example

Sampling error can occur when we use a sample of a large dataset to understand the complete dataset.

For example:

- Population → 1,000,000 customers
- Sample → 10,000 customers
- Population average purchase → ₹1,500
- Sample average purchase → ₹1,530

The ₹30 difference is sampling error.

Larger and well-selected samples generally help reduce sampling error.

```python
import numpy as np

# Create the population
population = np.array([100, 200, 300, 400, 500])

# Calculate the population mean
population_mean = np.mean(population)

# Select a sample
sample = np.array([100, 200, 500])

# Calculate the sample mean
sample_mean = np.mean(sample)

# Calculate sampling error
sampling_error = sample_mean - population_mean

print("Population Mean:", population_mean)
print("Sample Mean:", sample_mean)
print("Sampling Error:", sampling_error)
```

## Step 3 – Explanation

### What does the program do?

The program:

1. Creates a population.
2. Calculates the population mean.
3. Selects a smaller sample.
4. Calculates the sample mean.
5. Finds the difference between the two means.

### What statistical concept does it demonstrate?

It demonstrates **Sampling Error**.

The formula is:

**Sampling Error = Sample Mean − Population Mean**

### How do we interpret the output?

The population mean and sample mean are different because the sample contains only some members of the population.

The difference between them is the sampling error.

### Where is this used in AI/ML?

Sampling error is important when:

- Working with large datasets
- Creating training samples
- Performing statistical analysis
- Estimating population characteristics
- Evaluating model performance

Understanding sampling error helps us know how closely our sample represents the complete population.

# Sample Size

## Step 1 – Understand the Concept

**Sample Size** means the number of observations or members selected from a population for a study.

Choosing an appropriate sample size is important because:

- A very small sample may not represent the population well.
- A very large sample may require more time, cost, and computing resources.

### Simple Example

Suppose a company has **10,000 customers**.

Instead of collecting information from all 10,000 customers, the company may select **500 customers** for analysis.

Here:

- Population Size = 10,000
- Sample Size = 500

### General Idea

As the required accuracy and confidence increase, the required sample size generally increases.

A larger sample can help reduce sampling error, although sample size alone does not remove sampling bias.

## Step 2 – Demonstrate the Concept

### Example

Suppose a company wants to study customer satisfaction.

Population:

**10,000 customers**

The company chooses different sample sizes:

- Sample 1 → 50 customers
- Sample 2 → 500 customers
- Sample 3 → 1,000 customers

A larger sample provides more observations for analysis.

### AI/ML Example

Suppose an ML project has **1,000,000 customer records**.

Using all records may require more processing time.

A sample can be used for:

- Exploratory Data Analysis
- Testing data processing code
- Initial model experiments
- Understanding data patterns

After testing, the model can be trained using a larger dataset when required.

### Important Point

There is no single sample size that is correct for every problem.

Sample size depends on factors such as:

- Population size
- Desired confidence level
- Required precision
- Variability in the data
- Available resources

```python
import numpy as np

# Create a population of 10,000 customers
population = np.arange(1, 10001)

# Select different sample sizes
sample_50 = np.random.choice(population, size=50, replace=False)
sample_500 = np.random.choice(population, size=500, replace=False)
sample_1000 = np.random.choice(population, size=1000, replace=False)

# Display the sample sizes
print("Population Size:", len(population))
print("Sample 1 Size:", len(sample_50))
print("Sample 2 Size:", len(sample_500))
print("Sample 3 Size:", len(sample_1000))
```

## Step 3 – Explanation

### What does the program do?

The program:

1. Creates a population of 10,000 customers.
2. Creates three samples with different sizes.
3. Displays the size of each sample.

### What statistical concept does it demonstrate?

It demonstrates **Sample Size** — the number of observations selected from a population.

### How do we interpret the output?

The output shows:

- Population → 10,000 customers
- Sample 1 → 50 customers
- Sample 2 → 500 customers
- Sample 3 → 1,000 customers

The samples contain different numbers of observations from the same population.

### Where is this used in AI/ML?

Sample size is important for:

- Training ML models
- Testing models
- Exploratory Data Analysis
- Statistical analysis
- Survey and customer data analysis

A suitable sample size helps balance **accuracy, computation time, data collection effort, and cost**.
