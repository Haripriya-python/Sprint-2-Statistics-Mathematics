# Data Cleaning

## Step 1 – Understand the Concept

Real-world data may contain:

- Missing values
- Duplicate records
- Incorrect values

Machine learning models learn from the data we provide.

- Mean

# Missing Value Handling

## Step 1 – Understand the Concept

Missing values occur when some data is not available or not recorded in a dataset.

For example, a customer dataset may contain:

- Age → 25
- Income → 30000
- City → Missing

Machine Learning models generally need appropriate values to work with the data, so missing values should be identified and handled before model training.

### Common Ways to Handle Missing Values

1. Remove rows containing missing values
2. Replace missing values with the mean
3. Replace missing values with the median
4. Replace missing values with the mode
5. Use a suitable model-based method

### Role of Statistics

Statistics helps us choose an appropriate replacement value.

For example:

- Mean → useful when data is relatively balanced
- Median → useful when data contains outliers or is skewed
- Mode → useful for categorical values

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose we have customer ages:

[22, 25, 28, NaN, 30, 35]

One customer's age is missing.

We can calculate the median of the available ages and use it to replace the missing value.

This allows us to keep the record instead of removing the entire row.

### AI/ML Usage

Missing value handling is commonly used before:

- Model training
- Feature engineering
- Classification
- Regression
- Clustering

If missing values are not handled properly, they can affect statistical calculations and Machine Learning model performance.

```python
import pandas as pd
import numpy as np

# Create customer data with a missing value
data = {
    "Age": [22, 25, 28, np.nan, 30, 35],
    "Income": [25000, 30000, 35000, 40000, 45000, 50000]
}

df = pd.DataFrame(data)

# Display original data
print("Original Data:")
print(df)

# Calculate the median age
median_age = df["Age"].median()

# Fill the missing age with the median
df["Age"] = df["Age"].fillna(median_age)

print("\nMedian Age:", median_age)

print("\nData After Handling Missing Value:")
print(df)
```

## Step 3 – Explanation

### What does the program do?

The program:

1. Creates customer data.
2. Adds a missing value to the Age column.
3. Calculates the median age.
4. Replaces the missing age with the median.

### What statistical concept does it demonstrate?

It demonstrates **Missing Value Handling using the Median**.

The median is a statistical measure of central tendency and can be useful when the data may contain extreme values.

### How do we interpret the output?

The missing Age value is replaced with the calculated median.

This allows us to keep the complete row instead of deleting it.

### Where is this used in AI/ML?

Missing value handling is an important preprocessing step for:

- Classification
- Regression
- Clustering
- Feature Engineering
- Machine Learning model training

### Key Point

**Missing value handling helps prepare incomplete data for statistical analysis and Machine Learning.**

# Outlier Detection

## Step 1 – Understand the Concept

Outlier Detection is the process of identifying observations that are unusually far from most of the other values in a dataset.

An outlier may occur because of:

- Data entry errors
- Measurement errors
- Unusual events
- Genuine extreme observations

### Common Statistical Methods

Two common methods are:

1. **IQR Method**
2. **Z-Score Method**

### IQR Method

IQR stands for **Interquartile Range**.

IQR = Q3 - Q1

A common rule is:

**Lower Bound = Q1 - 1.5 × IQR**

**Upper Bound = Q3 + 1.5 × IQR**

Values outside these bounds are identified as potential outliers.

### Why is Outlier Detection Important?

Outliers can strongly affect:

- Mean
- Variance
- Standard deviation
- Statistical analysis
- Some Machine Learning models

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose a company records customer spending:

[1000, 1200, 1100, 1300, 1250, 15000]

Most customers spend around ₹1,000–₹1,300.

The value **₹15,000** is much larger than the other values.

It may be an outlier.

We can use the IQR method to identify whether ₹15,000 falls outside the normal range of the data.

### AI/ML Usage

Outlier detection is used in:

- Fraud detection
- Anomaly detection
- Sensor data analysis
- Financial analysis
- Data preprocessing

Outliers should not always be removed automatically. We should first understand whether the unusual value is an error or a genuine observation.

```python
import numpy as np

# Create customer spending data
spending = np.array([1000, 1200, 1100, 1300, 1250, 15000])

# Calculate Q1 and Q3
Q1 = np.percentile(spending, 25)
Q3 = np.percentile(spending, 75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliers = spending[
    (spending < lower_bound) | (spending > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

## Step 3 – Explanation

### What does the program do?

The program:

1. Creates customer spending data.
2. Calculates Q1 and Q3.
3. Calculates the IQR.
4. Calculates the lower and upper bounds.
5. Identifies values outside those bounds.

### What statistical concept does it demonstrate?

It demonstrates **Outlier Detection using the IQR Method**.

### How do we interpret the output?

If a value is below the lower bound or above the upper bound, it is identified as a potential outlier.

In this example, the spending value of **₹15,000** is much higher than the other values and is detected as an outlier.

### Where is this used in AI/ML?

Outlier detection is useful in:

- Fraud detection
- Anomaly detection
- Data preprocessing
- Financial datasets
- Sensor data
- Machine Learning pipelines

### Key Point

**Outlier detection helps identify unusual observations before they negatively affect statistical analysis or Machine Learning models.**

# Feature Scaling

## Step 1 – Understand the Concept

Feature Scaling is the process of bringing numerical features to a similar scale.

Different features can have very different ranges.

### Example

Suppose a dataset contains:

- Age → 18 to 60
- Income → ₹20,000 to ₹1,00,000

Income has much larger numerical values than Age.

Some Machine Learning algorithms can be affected by these differences in scale.

### Common Feature Scaling Methods

1. **Standardization**
2. **Min-Max Scaling**

### Standardization Formula

z = (x - mean) / standard deviation

After standardization, the feature generally has:

- Mean ≈ 0
- Standard deviation ≈ 1

### Why is Feature Scaling Used?

It can help algorithms that depend on:

- Distance
- Magnitude
- Gradient-based optimization

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose we are predicting customer behavior using:

- Age
- Annual Income

Example:

Age = 25

Income = ₹50,000

The numerical ranges are very different.

After standardization, both features are represented on a comparable scale.

### AI/ML Usage

Feature scaling is commonly important for:

- K-Nearest Neighbors (KNN)
- K-Means Clustering
- Support Vector Machines (SVM)
- Logistic Regression
- Neural Networks

Tree-based models such as Decision Trees and Random Forests generally do not require feature scaling in the same way.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Create sample customer data
data = {
    "Age": [20, 25, 30, 35, 40],
    "Income": [20000, 30000, 40000, 60000, 80000]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# Create StandardScaler
scaler = StandardScaler()

# Scale the features
scaled_data = scaler.fit_transform(df)

# Convert scaled data into a DataFrame
scaled_df = pd.DataFrame(
    scaled_data,
    columns=df.columns
)

print("\nScaled Data:")
print(scaled_df)
```

## Step 3 – Explanation

### What does the program do?

The program:

1. Creates customer data.
2. Contains Age and Income with different numerical ranges.
3. Applies Standardization using `StandardScaler`.
4. Converts the features to a comparable scale.

### What statistical concept does it demonstrate?

It demonstrates **Feature Scaling using Standardization**.

The transformation uses the mean and standard deviation of each feature.

### How do we interpret the output?

After scaling:

- Values are centered around 0.
- Features are represented on a similar scale.
- Positive values indicate observations above the feature mean.
- Negative values indicate observations below the feature mean.

### Where is this used in AI/ML?

Feature scaling is commonly used in:

- KNN
- K-Means
- SVM
- Logistic Regression
- Neural Networks

### Key Point

**Feature scaling prevents differences in feature magnitude from unnecessarily influencing algorithms that are sensitive to scale.**

# Data Normalization

## Step 1 – Understand the Concept

Data Normalization is the process of transforming numerical values to a common range.

A common method is **Min-Max Normalization**, which usually converts values to a range between **0 and 1**.

### Formula

Normalized Value = (x - minimum) / (maximum - minimum)

### Simple Example

Suppose the values are:

[10, 20, 30, 40, 50]

After Min-Max Normalization:

- 10 → 0
- 50 → 1
- Values between them are converted to values between 0 and 1.

### Why is Normalization Used?

Normalization can be useful when features have different ranges and an algorithm is sensitive to the scale of the input features.

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose we are analyzing customer data:

- Age → 20 to 60
- Annual Income → ₹20,000 to ₹1,00,000

The ranges are very different.

Using Min-Max Normalization, both features can be transformed to a similar range, such as 0 to 1.

### AI/ML Usage

Normalization can be useful for:

- K-Nearest Neighbors
- K-Means Clustering
- Neural Networks
- Distance-based algorithms

### Difference from Standardization

**Normalization:**
Usually scales values to a fixed range such as 0 to 1.

**Standardization:**
Transforms values based on mean and standard deviation.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Create customer data
data = {
    "Age": [20, 25, 30, 35, 40],
    "Income": [20000, 30000, 40000, 60000, 80000]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# Create Min-Max Scaler
scaler = MinMaxScaler()

# Normalize the data
normalized_data = scaler.fit_transform(df)

# Convert normalized data into a DataFrame
normalized_df = pd.DataFrame(
    normalized_data,
    columns=df.columns
)

print("\nNormalized Data:")
print(normalized_df)
```

## Step 3 – Explanation

### What does the program do?

The program:

1. Creates customer data.
2. Uses Age and Income as numerical features.
3. Applies Min-Max Normalization.
4. Converts the feature values to a range between 0 and 1.

### What statistical concept does it demonstrate?

It demonstrates **Data Normalization using Min-Max Scaling**.

### How do we interpret the output?

The smallest value of each feature becomes **0**.

The largest value becomes **1**.

The remaining values are converted proportionally between 0 and 1.

### Where is this used in AI/ML?

Normalization can be useful for:

- KNN
- K-Means Clustering
- Neural Networks
- Distance-based algorithms

### Key Point

**Normalization transforms numerical features to a common range, often 0 to 1, which can help scale-sensitive Machine Learning algorithms.**

# Model Evaluation

## Step 1 – Understand the Concept

Model Evaluation is the process of measuring how well a Machine Learning model performs on data.

After training a model, we need to check whether its predictions are accurate and reliable.

The evaluation metric depends on the type of Machine Learning problem.

### For Classification

Common metrics include:

- Accuracy
- Precision
- Recall
- F1-score

### For Regression

Common metrics include:

- Mean Absolute Error (MAE)
- Mean Squared Error (MSE)
- Root Mean Squared Error (RMSE)
- R² Score

### Role of Statistics

Statistical measures help us summarize the difference between actual values and predicted values.

They help us understand how well the model generalizes to unseen data.

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose a company wants to predict house prices.

Actual prices:

[20, 25, 30, 35]

Predicted prices:

[21, 24, 29, 36]

We can compare the actual and predicted values using evaluation metrics.

For example, **Mean Absolute Error (MAE)** calculates the average absolute difference between actual and predicted values.

### AI/ML Usage

Model evaluation is used to:

- Compare models
- Measure prediction performance
- Identify errors
- Select suitable models
- Check performance on unseen data

### Important Point

A model should be evaluated using data that was not used to train it, such as a test set.

```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Actual house prices
actual = np.array([20, 25, 30, 35])

# Predicted house prices
predicted = np.array([21, 24, 29, 36])

# Calculate evaluation metrics
mae = mean_absolute_error(actual, predicted)
mse = mean_squared_error(actual, predicted)
rmse = np.sqrt(mse)
r2 = r2_score(actual, predicted)

print("Mean Absolute Error:", mae)
print("Mean Squared Error:", mse)
print("Root Mean Squared Error:", rmse)
print("R² Score:", r2)
```

## Step 3 – Explanation

### What does the program do?

The program compares actual house prices with predicted house prices and calculates four regression evaluation metrics.

### What statistical concept does it demonstrate?

It demonstrates how statistical error measures can be used to evaluate a regression model.

### How do we interpret the output?

**MAE:**
Shows the average absolute difference between actual and predicted values. Lower values generally indicate smaller prediction errors.

**MSE:**
Calculates the average squared prediction error. Larger errors receive more weight because the errors are squared.

**RMSE:**
Is the square root of MSE and is expressed in the same units as the target variable.

**R² Score:**
Measures how much of the variation in the target is explained by the regression model. A value closer to 1 indicates a stronger fit for this metric.

### Where is this used in AI/ML?

Model evaluation is used in:

- Regression
- Classification
- Model comparison
- Model selection
- Production model monitoring

### Key Point

**Model evaluation uses statistical metrics to measure how well a Machine Learning model performs.**

# Linear Regression

## Step 1 – Understand the Concept

Linear Regression is a supervised Machine Learning algorithm used to predict a continuous numerical value.

It tries to find a relationship between an input feature and a target variable using a straight line.

### Basic Formula

ŷ = b₀ + b₁x

Where:

- ŷ = Predicted value
- b₀ = Intercept
- b₁ = Slope
- x = Input feature

### Simple Example

Suppose we want to predict a student's exam score based on study hours.

- Study Hours → Input feature
- Exam Score → Target

As study hours increase, exam scores may also increase.

Linear Regression tries to find the best-fit line representing this relationship.

### Role of Statistics

Linear Regression uses statistical concepts such as:

- Mean
- Variance
- Covariance
- Correlation
- Residuals
- Least Squares

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose a company wants to predict sales based on advertising spending.

| Advertising Spending | Sales |
|---|---:|
| ₹10,000 | ₹50,000 |
| ₹20,000 | ₹65,000 |
| ₹30,000 | ₹80,000 |
| ₹40,000 | ₹95,000 |

Here:

- Advertising Spending → Independent Variable
- Sales → Dependent Variable

Linear Regression learns the relationship between advertising spending and sales and uses it to predict future sales.

### AI/ML Usage

Linear Regression is used for:

- Sales prediction
- Price prediction
- Demand forecasting
- Revenue prediction
- Continuous value prediction

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Create study hours data
X = np.array([[1], [2], [3], [4], [5]])

# Create exam score data
y = np.array([50, 55, 65, 70, 80])

# Create the Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Predict score for 6 study hours
prediction = model.predict([[6]])

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
print("Predicted Score for 6 Hours:", prediction[0])
```

## Step 3 – Explanation

### What does the program do?

The program:

1. Creates study hours and exam score data.
2. Creates a Linear Regression model.
3. Trains the model using the sample data.
4. Predicts the exam score for 6 hours of study.

### What statistical concept does it demonstrate?

It demonstrates **Linear Regression**, which finds a best-fit line between an input variable and a continuous target.

The model estimates the slope and intercept from the training data.

### How do we interpret the output?

The **slope** represents the estimated change in exam score for one additional study hour.

The **intercept** represents the predicted score when study hours are zero according to the fitted line.

The predicted score gives the model's estimated exam score for 6 study hours.

### Where is this used in AI/ML?

Linear Regression is used for:

- Sales prediction
- House price prediction
- Demand forecasting
- Revenue prediction
- Other continuous-value prediction problems

### Key Point

**Linear Regression uses statistical relationships between variables to predict a continuous numerical value.**

# Logistic Regression

## Step 1 – Understand the Concept

Logistic Regression is a supervised Machine Learning algorithm mainly used for **classification problems**.

It predicts the probability of an observation belonging to a particular class.

For binary classification, the output is usually between **0 and 1**.

### Example

Suppose we want to predict whether a customer will purchase a product.

- 0 → Will not purchase
- 1 → Will purchase

The model first calculates a probability.

For example:

**Probability = 0.82**

If the classification threshold is 0.5:

**0.82 ≥ 0.5 → Class 1**

### Sigmoid Function

Logistic Regression uses the **sigmoid function** to convert a model score into a probability between 0 and 1.

### Role of Statistics

Logistic Regression uses statistical concepts such as:

- Probability
- Odds
- Log-odds
- Maximum likelihood
- Classification thresholds

## Step 2 – Demonstrate the Concept

### Real-World Example

A company wants to predict whether a customer will purchase a product based on the number of times they visit the website.

Example:

- 2 visits → likely not purchase
- 5 visits → uncertain
- 9 visits → likely purchase

The model learns the relationship between website visits and the purchase outcome.

### AI/ML Usage

Logistic Regression is commonly used for:

- Spam detection
- Customer purchase prediction
- Disease classification
- Fraud detection
- Binary classification

### Important Point

Despite its name, Logistic Regression is primarily used for **classification**, not continuous-value prediction.

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

# Number of website visits
X = np.array([[1], [2], [3], [4], [5], [6], [7], [8]])

# Purchase outcome
# 0 = No Purchase
# 1 = Purchase
y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

# Create Logistic Regression model
model = LogisticRegression()

# Train the model
model.fit(X, y)

# Predict probability for 6 website visits
probability = model.predict_proba([[6]])[0][1]

# Predict the class
prediction = model.predict([[6]])[0]

print("Probability of Purchase:", probability)
print("Predicted Class:", prediction)
```

## Step 3 – Explanation

### What does the program do?

The program:

1. Creates website visit data.
2. Uses purchase status as the target.
3. Trains a Logistic Regression model.
4. Calculates the probability of purchase for a customer with 6 website visits.
5. Predicts the final class.

### What statistical concept does it demonstrate?

It demonstrates **Logistic Regression**, which uses probability to perform classification.

### How do we interpret the output?

The model produces a probability between 0 and 1.

For example, if the output probability is:

**0.75**

the model estimates a 75% probability for the positive class.

Using the default threshold of 0.5:

- Probability ≥ 0.5 → Class 1
- Probability < 0.5 → Class 0

### Where is this used in AI/ML?

Logistic Regression is used for:

- Binary classification
- Spam detection
- Fraud detection
- Customer purchase prediction
- Medical classification

### Key Point

**Logistic Regression uses statistical probability to predict class membership.**

# Naive Bayes

## Step 1 – Understand the Concept

Naive Bayes is a supervised Machine Learning algorithm used mainly for **classification**.

It is based on **Bayes' Theorem** and calculates the probability of a class based on the available features.

It is called "Naive" because it assumes that the features are conditionally independent of each other given the class.

### Bayes' Theorem

P(A|B) = P(B|A) × P(A) / P(B)

Where:

- P(A|B) → Probability of A given B
- P(B|A) → Probability of B given A
- P(A) → Prior probability of A
- P(B) → Probability of B

### Simple Example

Suppose we want to classify an email as:

- Spam
- Not Spam

The model can use words appearing in the email as features and calculate the probability of each class.

### Role of Statistics

Naive Bayes uses:

- Probability
- Conditional probability
- Bayes' Theorem
- Prior probability
- Likelihood

## Step 2 – Demonstrate the Concept

### Real-World Example

An email contains words such as:

"free", "offer", and "prize".

A Naive Bayes classifier can calculate the probability that the email belongs to the Spam class based on the words it contains.

It compares the probabilities of different classes and assigns the email to the class with the appropriate predicted probability.

### AI/ML Usage

Naive Bayes is commonly used for:

- Spam detection
- Text classification
- Sentiment analysis
- Document classification
- News classification

### Key Point

Naive Bayes is particularly useful for many classification tasks involving text and categorical features.

```python
import numpy as np
from sklearn.naive_bayes import GaussianNB

# Create sample data
# Feature 1 = Study Hours
# Feature 2 = Practice Tests
X = np.array([
    [1, 0],
    [2, 1],
    [2, 0],
    [3, 1],
    [4, 2],
    [5, 2],
    [6, 3],
    [7, 3]
])

# Target
# 0 = Fail
# 1 = Pass
y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

# Create Naive Bayes model
model = GaussianNB()

# Train the model
model.fit(X, y)

# Predict for a new student
new_student = np.array([[5, 2]])

prediction = model.predict(new_student)
probability = model.predict_proba(new_student)

print("Predicted Class:", prediction[0])
print("Class Probabilities:", probability[0])
```

## Step 3 – Explanation

### What does the program do?

The program:

1. Creates sample student data.
2. Uses Study Hours and Practice Tests as features.
3. Uses Pass/Fail as the target.
4. Trains a Gaussian Naive Bayes model.
5. Predicts the class for a new student.
6. Displays the probability for each class.

### What statistical concept does it demonstrate?

It demonstrates **Naive Bayes classification**, which uses probability and Bayes' Theorem to classify observations.

### How do we interpret the output?

The model gives probabilities for the possible classes.

For example, if the probabilities are:

- Fail → 0.10
- Pass → 0.90

the model predicts **Pass**, because it has the higher predicted probability.

### Where is this used in AI/ML?

Naive Bayes is used for:

- Spam detection
- Text classification
- Sentiment analysis
- Document classification
- News classification

### Key Point

**Naive Bayes uses probability to classify data based on observed features.**

# Clustering

## Step 1 – Understand the Concept

Clustering is an **unsupervised Machine Learning technique** used to group similar data points together.

Unlike supervised learning, clustering does not require a target label for training.

The algorithm identifies patterns and creates groups based on the similarity between observations.

### Simple Example

Suppose an online store has customer data based on:

- Annual Income
- Spending Score

A clustering algorithm may group customers into different groups such as:

- Lower spending customers
- Medium spending customers
- Higher spending customers

### Role of Statistics

Clustering can involve statistical concepts such as:

- Mean
- Distance
- Variance
- Distribution
- Similarity

### Common Clustering Algorithms

- K-Means
- Hierarchical Clustering
- DBSCAN

## Step 2 – Demonstrate the Concept

### Real-World Example

A supermarket wants to understand its customers.

It collects:

- Annual Income
- Spending Score

Customers with similar income and spending behavior can be grouped together.

For example:

**Cluster 1:** Lower income and lower spending

**Cluster 2:** Medium income and medium spending

**Cluster 3:** Higher income and higher spending

The groups are discovered from the data rather than being provided as labels.

### AI/ML Usage

Clustering is used in:

- Customer segmentation
- Market analysis
- Recommendation systems
- Image analysis
- Anomaly detection
- Pattern discovery

```python
import pandas as pd
from sklearn.cluster import KMeans

# Create customer data
data = {
    "Annual_Income": [20, 22, 25, 45, 48, 50, 80, 85, 90],
    "Spending_Score": [20, 25, 18, 50, 55, 48, 80, 85, 78]
}

df = pd.DataFrame(data)

# Create K-Means model
model = KMeans(n_clusters=3, random_state=42, n_init=10)

# Create clusters
df["Cluster"] = model.fit_predict(df)

print("Customer Data with Clusters:")
print(df)

# Display cluster centers
print("\nCluster Centers:")
print(model.cluster_centers_)
```

## Step 3 – Explanation

### What does the program do?

The program:

1. Creates customer income and spending data.
2. Creates a K-Means clustering model with 3 clusters.
3. Groups customers based on their feature values.
4. Displays the cluster assigned to each customer.
5. Displays the center of each cluster.

### What statistical concept does it demonstrate?

It demonstrates **Clustering using K-Means**.

K-Means uses distances between observations and cluster centers to form groups.

### How do we interpret the output?

Customers with similar Annual Income and Spending Score are assigned to the same cluster.

The cluster centers represent the average position of the observations belonging to each cluster.

### Where is this used in AI/ML?

Clustering is used for:

- Customer segmentation
- Market analysis
- Recommendation systems
- Pattern discovery
- Unsupervised learning

### Key Point

**Clustering groups similar observations together without requiring predefined target labels.**

# Principal Component Analysis (PCA)

## Step 1 – Understand the Concept

Principal Component Analysis (PCA) is a statistical technique used for **dimensionality reduction**.

Dimensionality reduction means reducing the number of features while trying to preserve the important information in the data.

PCA transforms the original features into new variables called **Principal Components**.

### Simple Example

Suppose a dataset has:

- Height
- Weight
- Age
- Income
- Spending Score

Instead of using all 5 features, PCA can transform them into fewer components such as:

- Principal Component 1
- Principal Component 2

These components can represent much of the variation in the original data.

### Role of Statistics

PCA uses statistical concepts such as:

- Variance
- Covariance
- Correlation
- Eigenvalues
- Eigenvectors

### Why is PCA Used?

PCA can help to:

- Reduce the number of features
- Reduce computational complexity
- Visualize high-dimensional data
- Remove some redundant information
- Make datasets easier to analyze

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose a company has customer data with 10 different numerical features.

Working with all 10 features may make visualization and analysis difficult.

PCA can transform these features into fewer principal components, such as 2 or 3 components.

The reduced data can then be used for visualization or as input to another Machine Learning algorithm.

### AI/ML Usage

PCA is used for:

- Dimensionality reduction
- Data visualization
- Noise reduction
- Feature transformation
- Preprocessing
- Improving computational efficiency

### Important Point

PCA does not simply remove individual features. It creates new components from combinations of the original features.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Create sample customer data
data = {
    "Age": [20, 25, 30, 35, 40, 45],
    "Income": [20000, 30000, 40000, 50000, 60000, 70000],
    "Spending_Score": [30, 40, 50, 60, 70, 80]
}

df = pd.DataFrame(data)

# Standardize the features
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

# Apply PCA and reduce to 2 components
pca = PCA(n_components=2)
principal_components = pca.fit_transform(scaled_data)

# Create a DataFrame with the principal components
pca_df = pd.DataFrame(
    principal_components,
    columns=["Principal_Component_1", "Principal_Component_2"]
)

print("PCA Result:")
print(pca_df)

print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)
```

## Step 3 – Explanation

### What does the program do?

The program:

1. Creates customer data with three features.
2. Standardizes the features.
3. Applies PCA.
4. Reduces the three original features to two principal components.
5. Displays the explained variance ratio.

### What statistical concept does it demonstrate?

It demonstrates **Dimensionality Reduction using PCA**.

PCA finds new directions in the data that capture as much variance as possible.

### How do we interpret the output?

The **Explained Variance Ratio** tells us how much of the total variation in the original data is captured by each principal component.

For example:

- Principal Component 1 → captures the largest amount of variation
- Principal Component 2 → captures the next largest amount

The two components together represent the information retained after dimensionality reduction.

### Where is this used in AI/ML?

PCA is used for:

- Dimensionality reduction
- Data visualization
- Feature transformation
- Preprocessing
- Reducing computational complexity

### Key Point

**PCA transforms many original features into fewer principal components while retaining as much important variation as possible.**

# Recommendation Systems

## Step 1 – Understand the Concept

A Recommendation System is a Machine Learning system that suggests relevant products, movies, songs, or other items to users based on available data.

Statistics helps recommendation systems understand patterns in:

- User behavior
- Ratings
- Purchases
- Views
- Clicks
- Item similarity

### Simple Example

Suppose a user watches several action movies.

A recommendation system can analyze the user's previous activity and suggest other movies that may be relevant to that user.

### Role of Statistics

Statistical concepts can be used to:

- Analyze user behavior
- Calculate averages and ratings
- Measure similarity
- Identify patterns
- Estimate preferences

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose three users rate movies:

| User | Movie A | Movie B | Movie C |
|---|---:|---:|---:|
| User 1 | 5 | 4 | 1 |
| User 2 | 5 | 4 | 2 |
| User 3 | 1 | 2 | 5 |

User 1 and User 2 have similar rating patterns.

A recommendation system can identify this similarity and use User 2's preferences to recommend items to User 1.

### AI/ML Usage

Recommendation systems are used in:

- Netflix
- YouTube
- Amazon
- Spotify
- E-commerce applications

They can recommend:

- Movies
- Products
- Songs
- Videos
- Articles

```python
import pandas as pd

# Create a user-movie rating dataset
ratings = {
    "Movie_A": [5, 5, 1],
    "Movie_B": [4, 4, 2],
    "Movie_C": [1, 2, 5]
}

df = pd.DataFrame(
    ratings,
    index=["User_1", "User_2", "User_3"]
)

print("User Ratings:")
print(df)

# Calculate correlation between users
user_similarity = df.T.corr()

print("\nUser Similarity:")
print(user_similarity)
```

## Step 3 – Explanation

### What does the program do?

The program:

1. Creates a user-movie rating dataset.
2. Stores ratings given by different users.
3. Calculates the correlation between users.
4. Uses correlation to understand similarity in rating patterns.

### What statistical concept does it demonstrate?

It demonstrates how **correlation** can be used to measure similarity between users.

A high positive correlation means two users have similar rating patterns.

### How do we interpret the output?

If User 1 and User 2 have a high positive correlation, their movie preferences are similar in this dataset.

A recommendation system can use this type of similarity information to help identify items that may be relevant to a user.

### Where is this used in AI/ML?

Recommendation systems are used in:

- Movie recommendations
- Product recommendations
- Music recommendations
- Video recommendations
- Personalized content

### Key Point

**Recommendation systems use data about users and items to identify patterns and suggest relevant items.**
