# Topic 1 – Histogram

## Purpose

A Histogram is a statistical graph used to show the distribution of numerical data.

It groups numerical values into intervals called bins and shows how many observations fall into each interval.

## When to Use

- To understand the distribution of numerical data.
- To see how frequently values occur.
- To identify the center and spread of data.
- To identify possible skewness or unusual values.

## Advantages

- Easy to understand.
- Shows the shape of data distribution clearly.
- Helps identify the center and spread of the data.
- Useful for detecting skewness and possible outliers.

## Limitations

- It is mainly used for numerical data.
- The appearance can change depending on the number of bins.
- It does not show individual data values.
- It can be difficult to compare many datasets using separate histograms.

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt

# Create exam scores
scores = np.array([
    45, 50, 52, 55, 58,
    60, 62, 64, 65, 67,
    68, 70, 72, 74, 75,
    76, 78, 80, 82, 85
])

# Create histogram
plt.hist(scores, bins=5)

# Add labels and title
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Exam Scores")

# Display the histogram
plt.show()
```

## Code Explanation

### What the Program Does

The program creates a dataset containing exam scores of students.

It groups the scores into five intervals and displays the number of students in each interval using a histogram.

### Python Functions Used

- `np.array()` creates the numerical dataset.
- `plt.hist()` creates the histogram.
- `bins=5` divides the scores into five intervals.
- `plt.xlabel()` adds a label to the X-axis.
- `plt.ylabel()` adds a label to the Y-axis.
- `plt.title()` adds a title to the graph.
- `plt.show()` displays the graph.

### Statistical Concept Demonstrated

The histogram demonstrates how numerical data is distributed across different value ranges.

### Interpretation

The X-axis represents the exam score ranges.

The Y-axis represents the number of students whose scores fall within each range.

A taller bar means more students have scores in that particular range.

The histogram can help us understand whether the scores are concentrated in a particular range and whether the distribution appears symmetric or skewed.

### AI/ML Usage

Histograms are commonly used during Exploratory Data Analysis (EDA) to understand the distribution of numerical features before preprocessing and building a machine learning model.

# Topic 2 – Bar Chart

## Purpose

A Bar Chart is a graph used to compare values across different categories.

Each category is represented by a bar, and the length or height of the bar represents its value.

## When to Use

- To compare different categories.
- To display categorical data.
- To compare sales, products, departments, or regions.
- To identify the category with the highest or lowest value.

## Advantages

- Easy to read and understand.
- Makes category comparisons clear.
- Can display multiple categories.
- Useful for presenting business data.

## Limitations

- Not suitable for showing continuous data.
- Too many categories can make the chart difficult to read.
- It does not show the distribution of numerical data.
- Small differences between categories may be difficult to notice.


```python
# Import required library
import matplotlib.pyplot as plt

# Create product sales data
products = ["Laptop", "Mobile", "Tablet", "Headphones"]
sales = [120, 180, 90, 150]

# Create bar chart
plt.bar(products, sales)

# Add labels and title
plt.xlabel("Products")
plt.ylabel("Number of Units Sold")
plt.title("Product Sales")

# Display the bar chart
plt.show()
```

## Code Explanation

### What the Program Does

The program creates sales data for four different products.

It uses a bar chart to compare the number of units sold for each product.

### Python Functions Used

- `plt.bar()` creates the bar chart.
- `plt.xlabel()` adds a label to the X-axis.
- `plt.ylabel()` adds a label to the Y-axis.
- `plt.title()` adds a title to the graph.
- `plt.show()` displays the chart.

### Statistical Concept Demonstrated

The bar chart demonstrates the comparison of values across different categories.

### Interpretation

The X-axis represents the different products.

The Y-axis represents the number of units sold.

A taller bar represents a product with higher sales, while a shorter bar represents lower sales.

From the chart, Mobile has the highest number of units sold among the listed products.

### AI/ML Usage

Bar charts can be used during Exploratory Data Analysis (EDA) to understand categorical features and compare category frequencies before building a machine learning model.

# Topic 3 – Line Chart

## Purpose

A Line Chart is a graph used to show how a numerical value changes over a continuous sequence, especially over time.

It connects data points with lines to make trends and changes easy to identify.

## When to Use

- To show trends over time.
- To compare values across consecutive periods.
- To analyze sales, revenue, temperature, or website traffic.
- To identify increases, decreases, or patterns.

## Advantages

- Clearly shows trends and changes.
- Easy to compare values across time.
- Useful for time-series data.
- Makes increases and decreases easy to identify.

## Limitations

- Not suitable for unrelated categorical data.
- Too many lines can make the chart difficult to understand.
- Small changes may be difficult to notice.
- It may give a misleading impression if the scale of the axis is not appropriate.

```python
# Import required library
import matplotlib.pyplot as plt

# Create monthly sales data
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [200, 250, 230, 300, 350, 330]

# Create line chart
plt.plot(months, sales, marker="o")

# Add labels and title
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales Trend")

# Display the line chart
plt.show()
```

## Code Explanation

### What the Program Does

The program creates monthly sales data for six months.

It uses a line chart to show how sales change from January to June.

### Python Functions Used

- `plt.plot()` creates the line chart.
- `marker="o"` displays a marker at each data point.
- `plt.xlabel()` adds a label to the X-axis.
- `plt.ylabel()` adds a label to the Y-axis.
- `plt.title()` adds a title to the graph.
- `plt.show()` displays the chart.

### Statistical Concept Demonstrated

The line chart demonstrates how a numerical value changes across a continuous sequence, such as time.

### Interpretation

The X-axis represents the months.

The Y-axis represents the sales values.

The line shows the trend in sales over time.

Sales increase from January to February, decrease slightly in March, and then increase until May before decreasing slightly in June.

### AI/ML Usage

Line charts are useful during Exploratory Data Analysis (EDA) for understanding time-series data and identifying trends or patterns before applying forecasting or machine learning techniques.

# Topic 4 – Scatter Plot

## Purpose

A Scatter Plot is a graph used to show the relationship between two numerical variables.

Each point represents one observation, with one variable shown on the X-axis and another variable shown on the Y-axis.

## When to Use

- To identify relationships between two numerical variables.
- To check whether variables move together.
- To identify positive or negative relationships.
- To detect possible outliers.
- To visually explore correlation.

## Advantages

- Clearly shows the relationship between two variables.
- Helps identify positive, negative, or weak relationships.
- Makes possible outliers easy to notice.
- Useful for exploratory data analysis.

## Limitations

- It is mainly used for two numerical variables at a time.
- A scatter plot shows association, not causation.
- Large datasets can contain many overlapping points.
- It may be difficult to interpret when there is no clear pattern.


```python
# Import required library
import matplotlib.pyplot as plt

# Create study hours and exam scores
study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
exam_scores = [45, 50, 55, 60, 65, 70, 78, 85]

# Create scatter plot
plt.scatter(study_hours, exam_scores)

# Add labels and title
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")

# Display the scatter plot
plt.show()
```

## Code Explanation

### What the Program Does

The program creates data containing the number of hours students study and their corresponding exam scores.

It uses a scatter plot to visualize the relationship between study hours and exam scores.

### Python Functions Used

- `plt.scatter()` creates the scatter plot.
- `plt.xlabel()` adds a label to the X-axis.
- `plt.ylabel()` adds a label to the Y-axis.
- `plt.title()` adds a title to the graph.
- `plt.show()` displays the plot.

### Statistical Concept Demonstrated

The scatter plot demonstrates the relationship between two numerical variables.

### Interpretation

The X-axis represents study hours.

The Y-axis represents exam scores.

The points generally move upward as study hours increase, which indicates a positive relationship between study hours and exam scores in this example.

However, a scatter plot alone does not prove that studying more causes higher scores.

### AI/ML Usage

Scatter plots are commonly used during Exploratory Data Analysis (EDA) to understand relationships between numerical features and the target variable before building a machine learning model.

# Topic 5 – Pie Chart

## Purpose

A Pie Chart is a circular chart used to show how different categories contribute to a whole.

Each category is represented as a slice of the circle, and the size of the slice represents its proportion of the total.

## When to Use

- To show proportions or percentages.
- To display how a total value is divided among categories.
- To compare a small number of categories.
- To show market share, budget distribution, or expense categories.

## Advantages

- Easy to understand for a small number of categories.
- Clearly shows parts of a whole.
- Useful for displaying percentages.
- Simple and visually appealing.

## Limitations

- Difficult to compare similar-sized slices accurately.
- Not suitable for a large number of categories.
- Does not show trends over time.
- Can become difficult to read when many categories are included.


```python
# Import required library
import matplotlib.pyplot as plt

# Create expense data
categories = ["Food", "Travel", "Shopping", "Bills"]
expenses = [30, 20, 25, 25]

# Create pie chart
plt.pie(
    expenses,
    labels=categories,
    autopct="%1.1f%%"
)

# Add title
plt.title("Monthly Expense Distribution")

# Display the pie chart
plt.show()
```

## Code Explanation

### What the Program Does

The program creates monthly expense data for four categories.

It uses a pie chart to show how the total expenses are divided among these categories.

### Python Functions Used

- `plt.pie()` creates the pie chart.
- `labels` displays the category names.
- `autopct="%1.1f%%"` displays the percentage for each category.
- `plt.title()` adds a title to the chart.
- `plt.show()` displays the pie chart.

### Statistical Concept Demonstrated

The pie chart demonstrates how a total value is divided into different categories.

### Interpretation

Each slice represents one expense category.

The size of each slice represents that category's proportion of the total expenses.

Food represents 30% of the total, while Travel represents 20%. Shopping and Bills each represent 25%.

### AI/ML Usage

Pie charts can be used during Exploratory Data Analysis (EDA) to understand the proportion of categorical data, such as customer segments or class distributions.

However, for machine learning analysis, bar charts are often easier to compare when there are many categories.

# Topic 6 – Density Plot

## Purpose

A Density Plot is a graph used to show the distribution of numerical data using a smooth curve.

It represents the estimated probability density of different values in the dataset.

## When to Use

- To understand the shape of a numerical data distribution.
- To identify where values are concentrated.
- To compare the distributions of different groups.
- To identify skewness and possible multiple peaks.

## Advantages

- Shows the overall shape of the distribution clearly.
- Smoother than a histogram.
- Useful for comparing multiple distributions.
- Helps identify peaks and areas where data is concentrated.

## Limitations

- The shape depends on the smoothing method used.
- It may be difficult for beginners to interpret.
- It does not show individual observations.
- It can hide important details when the dataset is small.


```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt

# Create exam score data
scores = np.array([
    45, 50, 52, 55, 58,
    60, 62, 64, 65, 67,
    68, 70, 72, 74, 75,
    76, 78, 80, 82, 85
])

# Create density plot using a histogram with density=True
plt.hist(
    scores,
    bins=8,
    density=True,
    alpha=0.6
)

# Add labels and title
plt.xlabel("Exam Score")
plt.ylabel("Density")
plt.title("Density Distribution of Exam Scores")

# Display the plot
plt.show()
```

## Code Explanation

### What the Program Does

The program creates a dataset containing exam scores.

It uses a histogram with `density=True` to represent the probability density of the scores.

This gives a density-based view of how the values are distributed.

### Python Functions and Parameters Used

- `np.array()` creates the numerical dataset.
- `plt.hist()` creates the distribution plot.
- `bins=8` divides the data into eight intervals.
- `density=True` converts the histogram into a probability density representation.
- `alpha=0.6` controls the transparency of the bars.
- `plt.xlabel()` adds a label to the X-axis.
- `plt.ylabel()` adds a label to the Y-axis.
- `plt.title()` adds a title to the graph.
- `plt.show()` displays the plot.

### Statistical Concept Demonstrated

The plot demonstrates the probability density of numerical data.

Higher areas indicate regions where observations are more concentrated.

### Interpretation

The X-axis represents exam scores.

The Y-axis represents probability density.

The taller regions show where the exam scores are more concentrated.

The shape of the distribution helps us understand whether the data is concentrated around a particular range and whether it appears skewed.

### AI/ML Usage

Density plots are useful during Exploratory Data Analysis (EDA) to understand the distribution of numerical features and compare distributions before applying machine learning algorithms.

# Topic 7 – Box Plot

## Purpose

A Box Plot is a statistical graph used to summarize the distribution of numerical data.

It shows the median, quartiles, spread of the data, and possible outliers.

## When to Use

- To understand the spread of numerical data.
- To identify the median and quartiles.
- To detect possible outliers.
- To compare the distributions of different groups.
- To understand the variability of data.

## Advantages

- Shows the median and quartiles clearly.
- Helps identify possible outliers.
- Compact and easy to compare across groups.
- Useful for understanding data spread.

## Limitations

- Does not show the exact distribution shape in detail.
- Individual data points are not normally visible.
- Can be difficult for beginners to interpret.
- May hide patterns when datasets have different distribution shapes.


```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt

# Create employee salary data
salaries = np.array([
    25000, 27000, 28000, 30000, 31000,
    32000, 33000, 35000, 36000, 50000
])

# Create box plot
plt.boxplot(salaries)

# Add labels and title
plt.ylabel("Salary")
plt.title("Employee Salary Distribution")

# Display the box plot
plt.show()
```

## Code Explanation

### What the Program Does

The program creates a dataset containing employee salaries.

It uses a box plot to summarize the distribution of the salaries and identify possible unusual values.

### Python Functions Used

- `np.array()` creates the salary dataset.
- `plt.boxplot()` creates the box plot.
- `plt.ylabel()` adds a label to the Y-axis.
- `plt.title()` adds a title to the graph.
- `plt.show()` displays the plot.

### Statistical Concept Demonstrated

The box plot demonstrates the distribution of numerical data using quartiles and the median.

It can also help identify possible outliers.

### Interpretation

The box represents the middle 50% of the data, between the first quartile (Q1) and third quartile (Q3).

The line inside the box represents the median.

The whiskers show the typical range of the data according to the box plot rule.

Values plotted separately beyond the whiskers may be possible outliers.

In this example, the salary of 50,000 may appear as a possible high-value outlier compared with the other salaries.

### AI/ML Usage

Box plots are commonly used during Exploratory Data Analysis (EDA) to detect possible outliers and understand the spread of numerical features before preprocessing and model building.

# Topic 8 – Violin Plot

## Purpose

A Violin Plot is a statistical graph that combines information about the distribution and summary statistics of numerical data.

It shows the distribution shape using a density curve and usually includes the median and quartiles.

## When to Use

- To understand the distribution of numerical data.
- To compare distributions between different groups.
- To identify where values are concentrated.
- To observe the spread and shape of the data.

## Advantages

- Shows the distribution shape clearly.
- Provides more information than a basic box plot.
- Useful for comparing multiple groups.
- Shows areas where data is more concentrated.

## Limitations

- More difficult to interpret than a box plot.
- The shape depends on the density estimation method.
- Can be misleading with very small datasets.
- Individual observations are not always visible.


```python
# Import required libraries
import matplotlib.pyplot as plt

# Create salary data for three departments
development = [30000, 32000, 34000, 35000, 36000, 38000, 40000]
testing = [28000, 30000, 31000, 32000, 33000, 34000, 35000]
support = [25000, 27000, 28000, 29000, 30000, 31000, 32000]

# Combine the data
salary_data = [development, testing, support]

# Create violin plot
plt.violinplot(salary_data)

# Add labels
plt.xticks([1, 2, 3], ["Development", "Testing", "Support"])
plt.ylabel("Salary")
plt.title("Salary Distribution by Department")

# Display the violin plot
plt.show()
```

## Code Explanation

### What the Program Does

The program creates salary data for employees from three different departments.

It uses a violin plot to compare the distribution and spread of salaries across the departments.

### Python Functions Used

- `plt.violinplot()` creates the violin plot.
- `plt.xticks()` adds department names to the X-axis.
- `plt.ylabel()` adds a label to the Y-axis.
- `plt.title()` adds a title to the graph.
- `plt.show()` displays the plot.

### Statistical Concept Demonstrated

The violin plot demonstrates the distribution of numerical data using a density-based shape.

It helps show where values are concentrated and how widely the data is spread.

### Interpretation

The X-axis represents the three departments.

The Y-axis represents salary values.

The width of each violin shows where the data is more concentrated.

A wider section means more observations are concentrated around that salary range.

The shape of each violin can be compared to understand differences in salary distributions between departments.

### AI/ML Usage

Violin plots are useful during Exploratory Data Analysis (EDA) to compare the distribution of numerical features across different categories or groups before machine learning model development.

# Topic 9 – Pair Plot

## Purpose

A Pair Plot is a collection of scatter plots and distribution plots used to visualize relationships between multiple numerical variables in a dataset.

It helps us understand how different features are related to each other.

## When to Use

- To explore relationships between multiple numerical variables.
- To identify possible correlations.
- To observe patterns and clusters.
- To compare distributions of different features.
- During Exploratory Data Analysis (EDA).

## Advantages

- Shows relationships between many variables at the same time.
- Helps identify possible correlations.
- Makes patterns and clusters easier to notice.
- Useful for understanding a dataset before machine learning.

## Limitations

- Can become difficult to read when there are many variables.
- Not suitable for very large numbers of features.
- May require additional analysis to confirm relationships.
- It mainly focuses on pairwise relationships.


```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt

# Create a dataset
data = {
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8],
    "Attendance": [60, 65, 70, 75, 80, 85, 90],
    "Exam_Score": [50, 55, 60, 65, 72, 80, 88]
}

# Create DataFrame
df = pd.DataFrame(data)

# Create pair plot using Pandas
pd.plotting.scatter_matrix(
    df,
    figsize=(8, 8),
    diagonal="hist"
)

# Add title
plt.suptitle("Pair Plot of Student Data")

# Adjust layout
plt.tight_layout()

# Display the plot
plt.show()
```

## Code Explanation

### What the Program Does

The program creates a dataset containing study hours, attendance, and exam scores.

It uses a scatter matrix to visualize the relationships between all the numerical variables.

### Python Functions Used

- `pd.DataFrame()` creates a DataFrame from the dataset.
- `pd.plotting.scatter_matrix()` creates pairwise scatter plots.
- `diagonal="hist"` displays a histogram for each variable on the diagonal.
- `plt.suptitle()` adds a main title to the visualization.
- `plt.tight_layout()` adjusts the spacing of the plots.
- `plt.show()` displays the visualization.

### Statistical Concept Demonstrated

The visualization demonstrates relationships and distributions among multiple numerical variables.

The diagonal shows the distribution of each variable, while the other sections show relationships between pairs of variables.

### Interpretation

The visualization allows us to compare:

- Study Hours vs Attendance
- Study Hours vs Exam Score
- Attendance vs Exam Score

The points generally move upward together, indicating positive relationships between these variables in this example.

For example, as study hours increase, exam scores also generally increase.

### AI/ML Usage

Pair plots are commonly used during Exploratory Data Analysis (EDA) to understand relationships between features, identify possible correlations, and observe patterns before building a machine learning model.

# Topic 10 – Heatmap

## Purpose

A Heatmap is a graphical representation of data where different values are shown using different levels of color intensity.

It is commonly used to identify patterns, relationships, and correlations in a dataset.

## When to Use

- To visualize correlation between numerical variables.
- To identify strong and weak relationships.
- To compare values in a matrix or table.
- To find patterns in large datasets.
- During Exploratory Data Analysis (EDA).

## Advantages

- Makes patterns easy to identify visually.
- Useful for displaying correlation matrices.
- Can represent many values in a compact format.
- Helps identify strong and weak relationships quickly.

## Limitations

- Can become difficult to read with very large datasets.
- Color intensity may be difficult to interpret without a proper scale.
- It mainly shows relationships or values but does not explain why they occur.
- Correlation shown in a heatmap does not mean causation.

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt

# Create a dataset
data = {
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8],
    "Attendance": [60, 65, 70, 75, 80, 85, 90],
    "Assignment_Score": [50, 55, 60, 65, 72, 80, 88],
    "Exam_Score": [52, 57, 62, 67, 74, 82, 90]
}

# Create DataFrame
df = pd.DataFrame(data)

# Calculate correlation matrix
correlation = df.corr()

# Create heatmap using Matplotlib
plt.imshow(correlation, cmap="coolwarm")

# Add color scale
plt.colorbar()

# Add variable names
plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=45)
plt.yticks(range(len(correlation.columns)), correlation.columns)

# Add correlation values
for i in range(len(correlation.columns)):
    for j in range(len(correlation.columns)):
        plt.text(
            j,
            i,
            f"{correlation.iloc[i, j]:.2f}",
            ha="center",
            va="center"
        )

# Add title
plt.title("Correlation Heatmap")

# Adjust layout
plt.tight_layout()

# Display the heatmap
plt.show()
```

## Code Explanation

### What the Program Does

The program creates a dataset containing study hours, attendance, assignment scores, and exam scores.

It calculates the correlation between these numerical variables and displays the results using a heatmap.

### Python Functions Used

- `pd.DataFrame()` creates a DataFrame from the dataset.
- `df.corr()` calculates the correlation matrix.
- `plt.imshow()` displays the correlation matrix as a heatmap.
- `plt.colorbar()` adds a color scale to help interpret the values.
- `plt.xticks()` adds variable names to the X-axis.
- `plt.yticks()` adds variable names to the Y-axis.
- `plt.text()` displays the correlation values inside the cells.
- `plt.tight_layout()` adjusts the spacing.
- `plt.show()` displays the heatmap.

### Statistical Concept Demonstrated

The heatmap demonstrates the visualization of correlations between numerical variables.

Correlation values generally range from -1 to +1:

- A value close to +1 indicates a strong positive relationship.
- A value close to -1 indicates a strong negative relationship.
- A value close to 0 indicates a weak or no linear relationship.

### Interpretation

The rows and columns represent the numerical variables.

Each cell shows the correlation between two variables.

In this example, study hours and exam scores have a strong positive relationship because exam scores generally increase as study hours increase.

The diagonal values are 1 because each variable has a perfect positive correlation with itself.

### AI/ML Usage

Heatmaps are commonly used during Exploratory Data Analysis (EDA) to visualize feature correlations, identify highly related features, and support feature selection before building a machine learning model.
