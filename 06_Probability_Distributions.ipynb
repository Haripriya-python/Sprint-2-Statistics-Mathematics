# Topic 1 – Bernoulli Distribution

## Definition

Bernoulli Distribution is a probability distribution used for an experiment that has only two possible outcomes.

The two outcomes are usually represented as:

- Success = 1
- Failure = 0

The experiment is performed only once.

## Formula

P(X = x) = p^x × (1 - p)^(1-x)

Where:

- X = random variable
- x = 0 or 1
- p = probability of success
- 1 - p = probability of failure



## Where It Is Used

Bernoulli Distribution is used when an experiment has only two possible outcomes.

Examples:

- Customer purchases or does not purchase
- Machine works or fails
- Email is spam or not spam
- Loan is approved or rejected

## Real-World Example

Suppose a customer either purchases a product or does not purchase it.

If the probability of purchase is 0.7:

- Purchase = 1 → Probability = 0.7
- No Purchase = 0 → Probability = 0.3

Since there are only two possible outcomes, this is a Bernoulli experiment.



## AI/ML Example

Bernoulli Distribution is useful for binary classification problems where the target has two possible classes.

For example, an email classification model can predict:

- Spam = 1
- Not Spam = 0

Another example is customer churn prediction:

- Customer leaves = 1
- Customer stays = 0

The model works with a binary outcome, which can be represented using Bernoulli probabilities.

```python
# Python Visualization

import numpy as np
import matplotlib.pyplot as plt

# Set probability of success
p = 0.7

# Define possible outcomes
x = np.array([0, 1])

# Calculate probabilities
probabilities = np.array([
    1 - p,  # Failure probability
    p       # Success probability
])

# Create bar chart
plt.bar(x, probabilities)

# Add labels
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")

# Add outcome names
plt.xticks([0, 1], ["Failure (0)", "Success (1)"])

# Display the chart
plt.show()
```

## Python Code Explanation

### What the program does

The program creates a Bernoulli Distribution with a success probability of 0.7.

It calculates the probability of both possible outcomes:

- Failure = 0.3
- Success = 0.7

Then it displays these probabilities using a bar chart.

### What statistical concept it demonstrates

The program demonstrates Bernoulli Distribution, which represents a single experiment with two possible outcomes.

### How to interpret the output

The bar chart contains two bars:

- Failure (0) has a probability of 0.3.
- Success (1) has a probability of 0.7.

The success bar is higher because the probability of success is greater.

### Where it is used in AI/ML

Bernoulli Distribution can be used to represent binary outcomes in AI/ML.

Examples:

- Spam / Not Spam
- Fraud / Not Fraud
- Churn / Not Churn
- Purchase / No Purchase

### Python Visualization

The `matplotlib.pyplot.bar()` function is used to visualize the probability of the two outcomes.

The visualization makes it easy to compare the probability of success and failure.

# Topic 2 – Binomial Distribution

## Definition

Binomial Distribution is a probability distribution used to find the probability of getting a specific number of successes in a fixed number of independent trials.

Each trial has only two possible outcomes:

- Success = 1
- Failure = 0

The probability of success remains the same for every trial.

## Formula

P(X = k) = C(n, k) × p^k × (1 - p)^(n-k)

Where:

- n = total number of trials
- k = number of successes
- p = probability of success
- 1 - p = probability of failure
- C(n, k) = number of ways to choose k successes from n trials

## Where It Is Used

Binomial Distribution is used when:

- There is a fixed number of trials.
- Each trial has two possible outcomes.
- Trials are independent.
- The probability of success is the same for each trial.

## Real-World Example

Suppose a company sends 5 promotional emails.

Assume the probability that a customer opens an email is 0.6.

We can use Binomial Distribution to calculate the probability that exactly 3 out of 5 customers open the email.

Here:

- Number of trials (n) = 5
- Number of successes (k) = 3
- Probability of success (p) = 0.6

This helps the company estimate the number of successful customer responses.

## AI/ML Example

Binomial Distribution can be used when we want to model the number of successful outcomes from a fixed number of independent binary events.

For example, suppose an AI model classifies 10 emails as spam or not spam.

Each prediction can be considered as:

- Spam = Success
- Not Spam = Failure

We can use a Binomial Distribution to model the number of successful predictions when the probability of success is known or estimated.

Other examples include:

- Number of customers who click an advertisement
- Number of successful loan approvals
- Number of correct binary predictions

```python
# Topic 2 – Binomial Distribution
# Python Visualization

import numpy as np
import matplotlib.pyplot as plt
from math import comb

# Set number of trials
n = 10

# Set probability of success
p = 0.6

# Create possible number of successes
x = np.arange(0, n + 1)

# Calculate binomial probabilities
probabilities = np.array([
    comb(n, k) * (p ** k) * ((1 - p) ** (n - k))
    for k in x
])

# Create bar chart
plt.bar(x, probabilities)

# Add labels
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")

# Show the chart
plt.show()
```

## Python Code Explanation

### What the program does

The program calculates the probability of getting different numbers of successes in 10 independent trials.

The probability of success for each trial is set to 0.6.

It calculates the probability for 0, 1, 2, ..., up to 10 successes and displays the results using a bar chart.

### What statistical concept it demonstrates

The program demonstrates Binomial Distribution.

It shows how the probability is distributed across the possible number of successes in a fixed number of trials.

### How to interpret the output

The x-axis represents the number of successes.

The y-axis represents the probability of getting that number of successes.

For example, the bar above 6 represents the probability of getting exactly 6 successes out of 10 trials.

Since the probability of success is 0.6, values around 6 successes have higher probability.

### Where this concept is used in AI/ML

Binomial Distribution can be used to model the number of successes in repeated binary events.

Examples:

- Correct or incorrect predictions
- Customer clicks or does not click
- Spam or not spam
- Fraud or not fraud

### Python Visualization

`Matplotlib` is used to create the bar chart.

The chart makes it easier to understand how the probability changes as the number of successes increases.

# Topic 3 – Poisson Distribution

## Definition

Poisson Distribution is a probability distribution used to find the probability of a specific number of events occurring within a fixed interval of time or space.

It is useful when events happen independently and we are interested in how many times an event occurs.

The average rate of occurrence is represented by λ (lambda).

## Formula

P(X = k) = (e^(-λ) × λ^k) / k!

Where:

- X = number of events
- k = specific number of events
- λ = average number of events in the interval
- e = Euler's number, approximately 2.718
- k! = factorial of k

## Where It Is Used

Poisson Distribution is used when we want to model the number of times an event occurs during a fixed period or within a fixed area.

Examples:

- Number of customers arriving at a shop per hour
- Number of calls received by a call center per minute
- Number of website requests per second
- Number of defects in a product batch

## Real-World Example

Suppose a customer support center receives an average of 5 calls per hour.

We can use Poisson Distribution to find the probability of receiving exactly 3 calls in one hour.

Here:

- Average number of calls (λ) = 5
- Required number of calls (k) = 3

This helps the company understand the expected number of incoming calls and plan resources accordingly.

## AI/ML Example

Poisson Distribution can be used to model event counts in AI/ML applications.

For example, an online service can use it to model the number of user requests received by a server during a fixed time interval.

Other examples include:

- Number of website visits per minute
- Number of customer complaints per day
- Number of system errors per hour
- Number of events detected by a monitoring system

This can help in demand estimation, anomaly detection, and resource planning.

```python

# Python Visualization

import numpy as np
import matplotlib.pyplot as plt
from math import factorial, exp

# Set average number of events
lam = 5

# Create possible number of events
x = np.arange(0, 16)

# Calculate Poisson probabilities
probabilities = np.array([
    (exp(-lam) * (lam ** k)) / factorial(k)
    for k in x
])

# Create bar chart
plt.bar(x, probabilities)

# Add labels
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")

# Show the chart
plt.show()
```

## Python Code Explanation

### What the program does

The program calculates the probability of different numbers of events occurring when the average number of events is 5.

It calculates the probability for 0, 1, 2, ..., up to 15 events and displays the probabilities using a bar chart.

### What statistical concept it demonstrates

The program demonstrates Poisson Distribution.

It shows how the probability is distributed for the number of events occurring within a fixed interval when the average event rate is known.

### How to interpret the output

The x-axis represents the number of events.

The y-axis represents the probability of getting that number of events.

Here, the average number of events is 5.

The bars around 5 events have higher probabilities because 5 is the expected average number of events.

### Where this concept is used in AI/ML

Poisson Distribution can be used to model count-based events in AI/ML.

Examples:

- Server requests
- Website visits
- Customer complaints
- System errors
- Event occurrences

It can support demand prediction, monitoring, and anomaly detection.

### Python Visualization

Matplotlib is used to create the bar chart.

The visualization helps us understand how the probability changes for different numbers of events.

# Topic 4 – Uniform Distribution

## Definition

Uniform Distribution is a probability distribution in which all values within a given range have an equal chance of occurring.

It can be:

- Discrete Uniform Distribution – a fixed set of values have equal probability.
- Continuous Uniform Distribution – every value within a continuous interval has equal probability density.

Here, we focus on Continuous Uniform Distribution.

## Formula

For a continuous uniform distribution between a and b:

f(x) = 1 / (b - a)

Where:

- a = minimum value
- b = maximum value
- x = value between a and b

The probability density is constant throughout the interval.

## Where It Is Used

Uniform Distribution is used when every value within a specific range is considered equally likely.

Examples:

- Random number generation
- Simulation
- Random sampling
- Computer algorithms
- Game development

## Real-World Example

Suppose a bus arrives at a stop at a random time between 10:00 AM and 10:10 AM.

If we assume that every arrival time within these 10 minutes is equally likely, the waiting time can be represented using a Uniform Distribution.

For example, a passenger may wait anywhere between 0 and 10 minutes.

## AI/ML Example

Uniform Distribution is commonly used when generating random values for simulations and experiments.

It can also be used to initialize random parameters or create random samples within a specified range during machine learning experiments.

```python
# Python Visualization

import numpy as np
import matplotlib.pyplot as plt

# Set the minimum and maximum values
a = 0
b = 10

# Generate values between 0 and 10
x = np.linspace(a, b, 1000)

# Calculate uniform probability density
density = np.ones_like(x) / (b - a)

# Create the plot
plt.plot(x, density)

# Add labels
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.title("Uniform Distribution")

# Show the chart
plt.show()
```

## Python Code Explanation

### What the program does

The program creates a continuous Uniform Distribution between 0 and 10.

It generates values between 0 and 10 and calculates the same probability density for all these values.

The result is displayed using a line chart.

### What statistical concept it demonstrates

The program demonstrates Uniform Distribution.

It shows that the probability density is constant across the selected range.

### How to interpret the output

The x-axis represents values from 0 to 10.

The y-axis represents probability density.

The graph appears as a horizontal line because every value within the range has the same probability density.

Values outside the range are not included in this distribution.

### Where this concept is used in AI/ML

Uniform Distribution can be used in:

- Random data generation
- Simulation
- Random sampling
- Model parameter initialization
- Testing machine learning algorithms

For example, random values can be generated within a specific range when testing an ML model.

### Python Visualization

Matplotlib is used to create the line graph.

The horizontal shape of the graph visually demonstrates that the probability density is constant across the interval.

# Topic 5 – Normal Distribution

## Definition

Normal Distribution is a continuous probability distribution that is symmetric and has a bell-shaped curve.

In a normal distribution, most values are concentrated around the mean, while fewer values occur farther away from the mean.

The mean, median, and mode are equal in a perfectly normal distribution.

## Formula

The probability density function of Normal Distribution is:

f(x) = (1 / (σ√(2π))) × e^(-(x-μ)² / (2σ²))

Where:

- x = observed value
- μ = mean
- σ = standard deviation
- π = mathematical constant
- e = Euler's number

The mean determines the center of the distribution, while the standard deviation determines how spread out the distribution is.

## Where It Is Used

Normal Distribution is commonly used to model continuous measurements that tend to cluster around an average value.

Examples:

- Heights of people
- Measurement errors
- Exam scores in some settings
- Manufacturing measurements
- Sensor measurements

## Real-World Example

Suppose the heights of a large group of people are approximately normally distributed.

Most people will have heights close to the average height, while very short and very tall values will occur less frequently.

The data will form a bell-shaped curve around the mean.

## AI/ML Example

Normal Distribution is important in AI/ML because many statistical methods and machine learning techniques work with or assume approximately normally distributed features or errors.

It is also used in:

- Data preprocessing
- Statistical modeling
- Noise modeling
- Gaussian-based algorithms
- Standardization using mean and standard deviation

```python
# Python Visualization

import numpy as np
import matplotlib.pyplot as plt

# Set mean and standard deviation
mean = 50
std_dev = 10

# Create values around the mean
x = np.linspace(mean - 4 * std_dev, mean + 4 * std_dev, 1000)

# Calculate normal distribution density
density = (
    1 / (std_dev * np.sqrt(2 * np.pi))
) * np.exp(
    -((x - mean) ** 2) / (2 * std_dev ** 2)
)

# Create the plot
plt.plot(x, density)

# Add labels
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.title("Normal Distribution")

# Show the chart
plt.show()
```

## Python Code Explanation

### What the program does

The program creates a Normal Distribution with:

- Mean = 50
- Standard deviation = 10

It generates values around the mean and calculates the probability density for each value.

The result is displayed as a bell-shaped curve.

### What statistical concept it demonstrates

The program demonstrates Normal Distribution.

It shows how values are distributed around the mean and how the probability density decreases as we move farther away from the mean.

### How to interpret the output

The center of the curve is around 50 because the mean is 50.

The highest point of the curve occurs near the mean, showing that values close to 50 have higher probability density.

The curve becomes lower as values move away from the mean.

The standard deviation controls the spread of the curve:

- Smaller standard deviation → narrower curve
- Larger standard deviation → wider curve

### Where this concept is used in AI/ML

Normal Distribution is used in:

- Data preprocessing
- Feature standardization
- Statistical modeling
- Gaussian-based machine learning methods
- Modeling errors and noise

For example, standardization commonly transforms a feature using its mean and standard deviation so that it has mean 0 and standard deviation 1.

### Python Visualization

Matplotlib is used to draw the bell-shaped curve.

The visualization helps us understand the center, spread, and symmetric shape of a Normal Distribution.

# Topic 6 – Exponential Distribution

## Definition

Exponential Distribution is a continuous probability distribution used to model the waiting time until an event occurs.

It is commonly used when events occur continuously and independently at a constant average rate.

For example, it can represent the waiting time until the next customer arrives at a shop.

## Formula

The probability density function of Exponential Distribution is:

f(x) = λe^(-λx),  x ≥ 0

Where:

- x = waiting time
- λ (lambda) = average rate of events
- e = Euler's number, approximately 2.718

The mean waiting time is:

Mean = 1 / λ

## Where It Is Used

Exponential Distribution is used to model the time between events.

Examples:

- Waiting time for a customer to arrive
- Time between phone calls
- Time between website requests
- Time until a machine failure
- Time between system events

## Real-World Example

Suppose a customer support center receives an average of 2 calls per minute.

The Exponential Distribution can be used to model the waiting time until the next call arrives.

Here:

- Average rate (λ) = 2 calls per minute
- Average waiting time = 1 / 2 = 0.5 minutes

This helps the company understand expected waiting times and plan its resources.

## AI/ML Example

Exponential Distribution can be used in AI/ML to model waiting times and time-to-event data.

Examples include:

- Time between user requests
- Time until a system failure
- Time between customer interactions
- Time-to-event analysis

It can also be useful in simulations and reliability analysis.

```python
# Python Visualization

import numpy as np
import matplotlib.pyplot as plt

# Set the average event rate
lam = 2

# Create waiting time values
x = np.linspace(0, 5, 1000)

# Calculate exponential probability density
density = lam * np.exp(-lam * x)

# Create the plot
plt.plot(x, density)

# Add labels
plt.xlabel("Waiting Time")
plt.ylabel("Probability Density")
plt.title("Exponential Distribution")

# Show the chart
plt.show()
```

## Python Code Explanation

### What the program does

The program creates an Exponential Distribution with an event rate of 2 events per minute.

It calculates the probability density for different waiting times between 0 and 5 minutes.

The result is displayed using a line chart.

### What statistical concept it demonstrates

The program demonstrates Exponential Distribution.

It shows how the probability density changes as the waiting time increases.

### How to interpret the output

The x-axis represents the waiting time.

The y-axis represents probability density.

The curve starts at a high value when the waiting time is close to zero and decreases as the waiting time increases.

This means shorter waiting times have higher probability density than longer waiting times in this example.

### Where this concept is used in AI/ML

Exponential Distribution can be used in:

- Time-to-event analysis
- Reliability analysis
- System failure modeling
- Customer behavior analysis
- Simulation of waiting times

For example, an AI system can use waiting-time data to study how frequently events such as customer requests or system failures occur.

### Python Visualization

Matplotlib is used to create the line graph.

The decreasing curve visually shows how the probability density changes with increasing waiting time.
