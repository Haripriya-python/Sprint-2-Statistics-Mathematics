# Topic 1 – Mean

## 1. Definition

Mean is the average value of a set of numerical data.

It represents the central or typical value of the dataset.

## 2. Formula

Mean = Sum of all values / Number of values

Mathematically:

Mean = (x₁ + x₂ + x₃ + ... + xₙ) / n

Where:

- x₁, x₂, ..., xₙ = Data values
- n = Number of values



## 3. Numerical Example

Suppose the daily sales of a shop are:

₹2,000, ₹2,500, ₹3,000, ₹1,500, ₹2,000

### Calculation

Sum of values:

2000 + 2500 + 3000 + 1500 + 2000 = 11,000

Number of values = 5

Mean = 11,000 / 5

Mean = ₹2,200

```python
import numpy as np

# Create sales data
sales = np.array([2000, 2500, 3000, 1500, 2000])

# Calculate the mean
mean_sales = np.mean(sales)

print("Sales:", sales)
print("Mean Sales:", mean_sales)
```

# 4. Python Implementation – Explanation

- `numpy` is imported for numerical calculations.
- `np.array()` creates the sales dataset.
- `np.mean()` calculates the average of the sales values.
- The result is ₹2,200.

# 5. Business Example

A company wants to find its average monthly sales.

Monthly sales:

₹40,000, ₹45,000, ₹50,000, ₹55,000, ₹60,000

Mean = ₹50,000

The company can use the mean to understand its typical monthly sales.

# 6. AI/ML Example

Suppose a dataset contains customer ages:

25, 30, 35, 40, 45

Mean age = 35 years.

In AI/ML, mean can be used to understand the typical value of a numerical feature.

It can also be used to fill missing numerical values during data preprocessing.

# 7. Interpretation

The mean sales are ₹2,200.

This means that the average daily sales for the given 5 days are ₹2,200.

So, ₹2,200 represents the central or average level of the dataset.

### Important Point

Mean can be affected by extreme values or outliers.

For example, if one day's sales are extremely high, the mean will also increase.

# Topic 2 – Median

## 1. Definition

Median is the middle value of a dataset when the values are arranged in ascending or descending order.

If the dataset has an odd number of values, the middle value is the median.

If the dataset has an even number of values, the median is the average of the two middle values.

## 2. Formula

### For Odd Number of Values

Median = Middle value

### For Even Number of Values

Median = (Two middle values) / 2

Median is useful when the dataset contains extreme values or outliers because it is less affected by them than the mean.

## 3. Numerical Example

Consider the following data:

10, 20, 30, 40, 50

The values are already arranged in ascending order.

There are 5 values, so the middle value is:

Median = 30

Therefore, the median is 30.

### Even Number Example

Consider:

10, 20, 30, 40

There are 4 values.

The two middle values are 20 and 30.

Median = (20 + 30) / 2

Median = 25

```python
# Topic 2 – Median
# Step 3 – Implement the Concept

import numpy as np

# Create data
values = np.array([10, 20, 30, 40, 50])

# Calculate the median
median_value = np.median(values)

print("Values:", values)
print("Median:", median_value)
```

# 4. Python Implementation – Explanation

The program creates a NumPy array containing numerical values.

`np.median()` is used to calculate the middle value of the dataset.

The output is:

Median = 30

So, 30 is the middle value of the given dataset.

# 5. Business Example

A company wants to understand the typical salary of its employees.

Suppose the salaries are:

₹20,000, ₹25,000, ₹30,000, ₹35,000, ₹2,00,000

Median = ₹30,000

The very high salary of ₹2,00,000 does not affect the median significantly.

Therefore, median can give a better idea of the typical salary when extreme salaries are present.



# 6. AI/ML Example

Suppose a dataset contains the ages of customers:

20, 22, 25, 27, 80

Median = 25

The value 80 is an extreme value or outlier.

The median still represents the typical customer age better than the mean in this example.

In AI/ML, median can also be used to handle missing numerical values during data preprocessing.



# 7. Interpretation

The median is 30.

This means that 30 is the middle value of the given dataset when the values are arranged in order.

### Important Point

Median is less affected by extreme values or outliers compared with the mean.

Therefore, median is useful for datasets such as salary, house prices, and income where extreme values may exist.

### Library Used

NumPy is used because `np.median()` provides a simple and efficient way to calculate the median.

# Topic 3 – Mode

## 1. Definition

Mode is the value that appears most frequently in a dataset.

In simple words, mode tells us which value occurs the most number of times.

A dataset can have:

- One mode → Unimodal
- Two modes → Bimodal
- More than two modes → Multimodal
- No mode → When all values occur only once

## 2. Formula

There is no specific mathematical formula for mode.

Mode = The value with the highest frequency

### Example

Data:

10, 20, 20, 30, 40

20 appears two times, while the other values appear only once.

Therefore:

Mode = 20

## 3. Numerical Example

Consider the following data:

10, 20, 20, 30, 40, 20, 50

### Frequency of Values

- 10 → 1 time
- 20 → 3 times
- 30 → 1 time
- 40 → 1 time
- 50 → 1 time

The value 20 appears the most frequently.

Therefore:

Mode = 20


```python
# Topic 3 – Mode
# Step 3 – Implement the Concept

import pandas as pd

# Create data
values = [10, 20, 20, 30, 40, 20, 50]

# Calculate the mode
mode_value = pd.Series(values).mode()

print("Values:", values)
print("Mode:", mode_value.tolist())
```

## 4. Python Implementation – Explanation

The program creates a list of numerical values.

`pd.Series()` converts the list into a Pandas Series.

`.mode()` identifies the value that occurs most frequently.

The output is:

Mode = 20

Therefore, 20 is the most frequently occurring value in the dataset.


# 5. Business Example

The most frequently purchased size is:

### Important Point

Mode is different from mean and median.

- Mean → Average value

Mode is particularly useful for categorical data such as product type, payment method, department, color, or customer preference.

## 1. Definition

Weighted Mean is an average where each value is given a different level of importance or weight.

In a normal mean, all values have equal importance.

In a weighted mean, some values contribute more to the final average than others.

## 2. Formula

Weighted Mean = Σ(w × x) / Σw

Where:

- x = Data value
- w = Weight assigned to the value
- Σ(w × x) = Sum of value × weight
- Σw = Sum of all weights



## 3. Numerical Example

Suppose a student's final grade is calculated using:

- Assignment = 80 marks, Weight = 20%
- Exam = 90 marks, Weight = 80%

### Calculation

Weighted Mean = (80 × 0.20 + 90 × 0.80) / (0.20 + 0.80)

Weighted Mean = (16 + 72) / 1

Weighted Mean = 88

Therefore:

Weighted Mean = 88

The exam has a higher weight, so it has a greater effect on the final result.

```python
import numpy as np

# Create values and their weights
marks = np.array([80, 90])
weights = np.array([0.20, 0.80])

# Calculate weighted mean
weighted_mean = np.average(marks, weights=weights)

print("Marks:", marks)
print("Weights:", weights)
print("Weighted Mean:", weighted_mean)
```



## 4. Python Implementation – Explanation

`marks` contains the values.

`weights` contains the importance assigned to each value.

`np.average()` calculates the weighted mean using the given weights.

### Output

Weighted Mean = 88.0

### Statistical Concept

The program calculates an average where each value contributes according to its assigned weight.

### Library Used

NumPy is used because `np.average()` supports weighted average calculations.



## 5. Business Example

A company evaluates an employee using:

- Project Performance = 80, Weight = 60%
- Attendance = 90, Weight = 40%

The project performance has a higher weight.

Weighted Mean:

(80 × 0.60) + (90 × 0.40)

= 48 + 36

= 84

Therefore, the employee's weighted performance score is 84.

The company can use weighted mean when different performance factors have different levels of importance.


## 6. AI/ML Example

Suppose an ML model evaluates two types of errors:

- Common class error = 80, Weight = 0.3
- Important class error = 90, Weight = 0.7

The second class has a higher weight because it is more important for the application.

Weighted Mean:

(80 × 0.3) + (90 × 0.7)

= 24 + 63

= 87

Weighted averages can be useful in AI/ML when different samples, classes, or measurements need different levels of importance.



## 7. Interpretation

The weighted mean is 88.

This means the overall average is 88 after considering the different importance of the values.

The exam has a weight of 80%, so it contributes more to the final result than the assignment.

### Important Point

Weighted Mean is useful when all values do not have equal importance.

For example:

- Exam scores
- Employee performance
- Product ratings
- Investment returns
- AI/ML weighted data

Weighted Mean gives more influence to values with higher weights.

# Topic 5 – Geometric Mean

## 1. Definition

Geometric Mean is a type of average used when values are multiplied together or when data represents growth rates, ratios, or percentages.

It is calculated by multiplying all the values and then taking the nth root, where n is the number of values.

## 2. Formula

Geometric Mean = (x₁ × x₂ × x₃ × ... × xₙ)^(1/n)

Where:

- x₁, x₂, ..., xₙ = Data values
- n = Number of values

Geometric Mean is generally used with positive values.



## 3. Numerical Example

Consider the following values:

2, 4, 8

### Step 1 – Multiply the values

2 × 4 × 8 = 64

### Step 2 – Find the cube root

There are 3 values, so:

Geometric Mean = 64^(1/3)

Geometric Mean = 4

Therefore:

Geometric Mean = 4

```python
import numpy as np

# Create data
values = np.array([2, 4, 8])

# Calculate geometric mean
geometric_mean = np.prod(values) ** (1 / len(values))

print("Values:", values)
print("Geometric Mean:", geometric_mean)
```

## 4. Python Implementation – Explanation

`np.prod()` multiplies all the values in the dataset.

`len(values)` gives the number of values.

The result is raised to the power of `1/n` to calculate the geometric mean.

### Output

Geometric Mean = 4.0

### Statistical Concept

The program calculates the geometric mean by multiplying all values and taking the nth root.

### Library Used

NumPy is used for numerical calculations such as multiplication and mathematical operations.



## 5. Business Example

Suppose a company's sales grow by different factors over three years:

Year 1 → 1.10  
Year 2 → 1.20  
Year 3 → 1.30

These values represent multiplicative growth factors.

The geometric mean can be used to find the average growth factor over the three years.

Geometric Mean ≈ 1.196

This represents an average growth factor of approximately 1.196 per year.



## 6. AI/ML Example

Suppose an AI system evaluates improvement ratios across different stages:

1.2, 1.5, 1.8

The geometric mean can be used to summarize the overall multiplicative improvement across the stages.

Geometric mean is useful when values represent ratios, growth factors, or multiplicative changes.

It can also be used in areas such as model performance analysis and growth-based measurements.



## 7. Interpretation

The geometric mean of 2, 4, and 8 is 4.

This means that 4 represents the central multiplicative value of the dataset.

Geometric Mean is particularly useful for:

- Growth rates
- Ratios
- Percentages
- Investment returns
- Multiplicative changes

### Important Point

Geometric Mean is generally used with positive values.

It is different from arithmetic mean because it is based on multiplication rather than addition.

# Topic 6 – Harmonic Mean

## 1. Definition

Harmonic Mean is a type of average that is mainly used for rates, ratios, and values that represent a "per unit" relationship.

It gives more importance to smaller values compared with the arithmetic mean.

## 2. Formula

Harmonic Mean = n / Σ(1/x)

Where:

- x = Data values
- n = Number of values
- Σ(1/x) = Sum of the reciprocals of all values

For two values:

Harmonic Mean = 2 / (1/x₁ + 1/x₂)

Harmonic Mean is generally used with positive, non-zero values.



## 3. Numerical Example

Suppose a vehicle travels at:

60 km/h and 40 km/h

### Calculation

Harmonic Mean = 2 / (1/60 + 1/40)

Harmonic Mean = 2 / (0.0167 + 0.025)

Harmonic Mean ≈ 48 km/h

Therefore:

Harmonic Mean ≈ 48 km/h

The harmonic mean is useful here because speed is a rate.

```python


import numpy as np

# Create speed data
speeds = np.array([60, 40])

# Calculate harmonic mean
harmonic_mean = len(speeds) / np.sum(1 / speeds)

print("Speeds:", speeds)
print("Harmonic Mean:", harmonic_mean)
```


## 4. Python Implementation – Explanation

`1 / speeds` calculates the reciprocal of each speed.

`np.sum()` adds the reciprocal values.

The formula `n / Σ(1/x)` is then used to calculate the harmonic mean.

### Output

Harmonic Mean ≈ 48.0 km/h

### Statistical Concept

The program calculates the average of rate values using the harmonic mean.

### Library Used

NumPy is used for numerical calculations and array operations.



## 5. Business Example

A delivery company measures the delivery rate of a vehicle on two different routes:

- Route 1 = 60 km/h
- Route 2 = 40 km/h

The harmonic mean is approximately:

48 km/h

The company can use this value to represent the average speed across routes when the rates are being combined appropriately.


## 6. AI/ML Example

In classification problems, the F1-score is the harmonic mean of Precision and Recall.

Formula:

F1 Score = 2 × (Precision × Recall) / (Precision + Recall)

Suppose:

Precision = 0.80
Recall = 0.60

F1 Score = 2 × (0.80 × 0.60) / (0.80 + 0.60)

F1 Score ≈ 0.686

The F1-score combines Precision and Recall into a single measurement and is useful when both are important.

## 7. Interpretation

The harmonic mean of 60 km/h and 40 km/h is approximately 48 km/h.

This means that 48 km/h represents the combined average rate using the harmonic mean.

### Important Point

Harmonic Mean is especially useful for:

- Speed
- Rates
- Ratios
- Performance rates
- Precision and Recall in AI/ML

It gives more influence to smaller values than the arithmetic mean.
