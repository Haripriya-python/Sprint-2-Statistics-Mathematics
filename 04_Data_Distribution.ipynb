# 1. Normal Distribution

## Step 1 – Understand the Concept

A Normal Distribution is a probability distribution where most of the data values are concentrated around the mean, and fewer values appear as we move farther away from the mean.

It has a bell-shaped curve and is symmetric around the mean.

In a normal distribution:

- Mean, median, and mode are equal.
- Most values are close to the mean.
- Very few values are found at the extreme ends.
- The distribution is symmetric on both sides of the mean.

A normal distribution is described mainly using:

- Mean (μ) – represents the center of the distribution.
- Standard deviation (σ) – represents how spread out the data is.

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose we collect the heights of a large group of people.

Most people may have heights close to the average height, while very short and very tall people are less common.

For example:

150, 155, 158, 160, 162, 165, 168, 170, 175

When many observations are collected, the values can form a bell-shaped pattern around the average.

### Numerical Example

Suppose:

Mean = 100
Standard Deviation = 10

Values around 100 will be more common than values far away from 100.

### AI/ML Use Case

Normal distributions are commonly used in statistics and machine learning for understanding numerical data, probability, statistical analysis, and some machine learning algorithms that make assumptions about the distribution of data.

```python
import numpy as np
import matplotlib.pyplot as plt

# Set the mean and standard deviation
mean = 100
std = 10

# Create values for the normal distribution
x = np.linspace(60, 140, 1000)

# Calculate the normal distribution values
y = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(
    -0.5 * ((x - mean) / std) ** 2
)

# Plot the normal distribution
plt.plot(x, y)

# Add labels and title
plt.xlabel("Values")
plt.ylabel("Probability Density")
plt.title("Normal Distribution")

# Display the graph
plt.show()
```

## Step 3 – Explanation

### What does the program do?

The program creates a normal distribution using a mean of 100 and a standard deviation of 10.

It generates values between 60 and 140 and calculates the corresponding probability density for each value.

Finally, it plots the values to create the bell-shaped normal distribution curve.

### What statistical concept does it demonstrate?

The program demonstrates the shape and characteristics of a Normal Distribution.

The highest point of the curve is around the mean, and the curve decreases as we move away from the mean.

### How do we interpret the output?

The graph should look like a bell-shaped curve.

The center of the curve is around 100 because the mean is 100.

Most of the probability density is concentrated around the mean, while the values become less likely as they move farther away from the mean.

The standard deviation controls the spread of the curve.

### Where is this concept used in AI/ML?

Normal Distribution is used in statistical analysis, probability calculations, feature analysis, hypothesis testing, and machine learning techniques that assume or work with approximately normally distributed data.

# 2. Uniform Distribution

## Step 1 – Understand the Concept

A Uniform Distribution is a probability distribution where all values within a given range have an equal chance of occurring.

In a continuous uniform distribution, every value between the minimum and maximum limits has the same probability density.

For example, if a random number is selected between 1 and 10, values within that range have an equal chance of being selected.

A uniform distribution is usually represented by a rectangular or flat-shaped graph.

Important points:

- It has a minimum and maximum limit.
- Values within the range have equal probability density.
- The distribution has a flat shape.

## Step 2 – Demonstrate the Concept

### Numerical Example

Suppose a random number is generated between 1 and 10.

The possible values are within the range:

Minimum = 1
Maximum = 10

If the random process follows a uniform distribution, there is no preference for values near 1, 5, or 10.

### Real-World Example

Suppose a system randomly selects a number between 1 and 100 for testing.

If every number has an equal chance of being selected, the selection follows a uniform distribution.

### AI/ML Use Case

Uniform distributions can be used for random data generation, simulations, testing, and initializing certain parameters in machine learning algorithms.

```python
import numpy as np
import matplotlib.pyplot as plt

# Set the minimum and maximum values
minimum = 1
maximum = 10

# Generate random values from a uniform distribution
data = np.random.uniform(minimum, maximum, 1000)

# Plot the distribution
plt.hist(data, bins=20)

# Add labels and title
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")

# Display the graph
plt.show()
```

## Step 3 – Explanation

### What does the program do?

The program generates 1000 random values between 1 and 10 using NumPy's uniform distribution.

It then displays the generated values using a histogram.

### What statistical concept does it demonstrate?

The program demonstrates Uniform Distribution, where values are randomly generated within a specified range with equal probability density.

### How do we interpret the output?

The histogram should show a roughly flat distribution across the range from 1 to 10.

Because the values are randomly generated, the bars may not have exactly the same height, but they should be approximately similar when a large number of values are generated.

### Where is this concept used in AI/ML?

Uniform Distribution can be used for:

- Random data generation
- Simulation
- Model testing
- Random sampling
- Parameter initialization in some machine learning methods

# 3. Gaussian Distribution

## Step 1 – Understand the Concept

A Gaussian Distribution is another name for the Normal Distribution.

It is a continuous probability distribution with a symmetric, bell-shaped curve.

The center of the distribution is determined by the mean, while the spread is controlled by the standard deviation.

In a Gaussian Distribution:

- Mean, median, and mode are equal.
- The curve is symmetric around the mean.
- Most observations are close to the mean.
- Fewer observations occur farther away from the mean.
- The shape depends on the mean and standard deviation.

Gaussian Distribution is widely used in statistics because many real-world measurements can be approximately represented by this distribution.

## Step 2 – Demonstrate the Concept

### Numerical Example

Suppose the marks of students have:

Mean = 60
Standard Deviation = 10

Most student marks will be around 60, while fewer students will have marks much lower or much higher than 60.

### Real-World Example

Measurements such as height, measurement errors, and some types of test scores can approximately follow a Gaussian distribution under suitable conditions.

### AI/ML Use Case

Gaussian distributions are used in machine learning for probability modelling, statistical inference, noise modelling, and algorithms such as Gaussian Naive Bayes.

```python
import numpy as np
import matplotlib.pyplot as plt

# Set the mean and standard deviation
mean = 60
std = 10

# Create values around the mean
x = np.linspace(20, 100, 1000)

# Calculate the Gaussian probability density
y = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(
    -0.5 * ((x - mean) / std) ** 2
)

# Plot the Gaussian distribution
plt.plot(x, y)

# Add labels and title
plt.xlabel("Values")
plt.ylabel("Probability Density")
plt.title("Gaussian Distribution")

# Display the graph
plt.show()
```

## Step 3 – Explanation

### What does the program do?

The program creates a Gaussian Distribution using a mean of 60 and a standard deviation of 10.

It calculates the probability density for different values and plots them as a curve.

### What statistical concept does it demonstrate?

It demonstrates the bell-shaped Gaussian Distribution and shows how the mean and standard deviation affect the distribution.

### How do we interpret the output?

The graph should have a bell-shaped curve.

The highest point is around 60 because the mean is 60.

The curve decreases on both sides as the values move farther away from the mean.

### Where is this concept used in AI/ML?

Gaussian Distribution is used in:

- Probability modelling
- Statistical inference
- Noise modelling
- Gaussian Naive Bayes
- Data analysis
- Some machine learning algorithms that assume normally distributed data

### Important Note

Gaussian Distribution and Normal Distribution refer to the same probability distribution. The term "Gaussian" comes from the mathematician Carl Friedrich Gauss.

# Right Skew

## Step 1 – Understand the Concept

Right Skew, also called **Positive Skewness**, occurs when the distribution has a **longer tail on the right side**.

In a right-skewed distribution:

- Most values are concentrated on the left side.
- A few unusually large values extend the distribution toward the right.
- Usually, **Mean > Median > Mode**.
- The long right tail is caused by higher-value observations.

### Simple Example

Consider employee salaries:

`20,000, 22,000, 23,000, 25,000, 27,000, 80,000`

Most salaries are between ₹20,000 and ₹27,000, but one very high salary of ₹80,000 creates a **long tail on the right**.

This is called a **Right-Skewed Distribution**.

## Step 2 – Demonstrate the Concept

Imagine the following distribution:

                    *
                    *
                *   *
                *   *
            *   *   *
        *   *   *   *
    *   *   *   *   *   *   *   →
--------------------------------
                         Right Tail

Most observations are on the left, while a few larger values extend toward the right.

### Real-World Examples

- Income and salary data
- House prices
- Hospital expenses
- Customer spending
- Online purchase amounts

These datasets can contain a small number of very large values, creating a right tail.

```python
import numpy as np
import matplotlib.pyplot as plt

# Create a right-skewed dataset
data = np.random.exponential(scale=2, size=1000)

# Plot the distribution
plt.hist(data, bins=30)

# Add labels and title
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Right-Skewed Distribution")

# Display the graph
plt.show()
```

## Step 3 – Explanation

### What does the program do?

The program creates 1,000 values using an exponential distribution and displays them as a histogram.

The histogram helps us visually identify the shape of the data distribution.

### What statistical concept does it demonstrate?

It demonstrates **Right Skew (Positive Skewness)**.

The majority of values are concentrated on the left, while a smaller number of larger values create a long tail toward the right.

### How do we interpret the output?

When looking at the histogram:

- Most values appear on the left side.
- There are fewer large values.
- The distribution has a longer tail toward the right.
- This indicates a **right-skewed distribution**.

### Where is this used in AI/ML?

Right-skewed data is common in real-world datasets such as:

- Income data
- Customer spending
- Property prices
- Medical costs
- Transaction amounts

Understanding skewness helps in **data analysis and preprocessing**. In some ML models, highly skewed features may be transformed using techniques such as a **log transformation** to make the data more suitable for modelling.

# Symmetric Distribution

## Step 1 – Understand the Concept

A **Symmetric Distribution** is a distribution where the values are distributed in a balanced way around the center.

If we draw a vertical line through the center of the distribution, both sides look approximately the same.

In a symmetric distribution:

- The left side and right side have a similar shape.
- The center is balanced.
- Usually, **Mean ≈ Median ≈ Mode**.
- There is no strong left or right skew.

### Simple Example

Consider:

`10, 20, 30, 40, 50, 60, 70`

The values are balanced around **40**.

There are equal distances on both sides of the center:

`10  20  30  |40|  50  60  70`

## Step 2 – Demonstrate the Concept

A symmetric distribution can look like this:

                    *
                  *   *
                *       *
              *           *
            *               *
          *                   *
        *                       *
-----------------------------------
              Center

The left and right sides have a similar shape.

### Real-World Example

A dataset of exam scores may approximately follow a symmetric distribution when most students score around the average and very few students score extremely low or extremely high.

```python
import numpy as np
import matplotlib.pyplot as plt

# Create a symmetric dataset
data = np.random.normal(loc=50, scale=10, size=1000)

# Plot the distribution
plt.hist(data, bins=30)

# Add labels and title
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Symmetric Distribution")

# Display the graph
plt.show()
```

## Step 3 – Explanation

### What does the program do?

The program generates 1,000 values using a normal distribution with a mean of 50 and a standard deviation of 10.

It then displays the values using a histogram.

### What statistical concept does it demonstrate?

It demonstrates a **Symmetric Distribution**.

The generated values are distributed approximately equally around the center.

### How do we interpret the output?

From the histogram:

- Most values are around 50.
- The left and right sides have a similar shape.
- There is no strong tail on either side.
- This indicates that the distribution is approximately symmetric.

### Where is this used in AI/ML?

Symmetric distributions are useful when analyzing numerical features in datasets.

They are commonly considered during:

- Exploratory Data Analysis (EDA)
- Feature analysis
- Statistical modelling
- Data preprocessing
- Machine learning model preparation

Understanding whether a feature is symmetric or skewed helps us decide whether any data transformation may be useful.

# Kurtosis

## Step 1 – Understand the Concept

**Kurtosis** is a statistical measure that describes the **shape of a distribution**, especially the heaviness of its tails and the presence of extreme values.

It helps us understand whether a dataset has more or fewer extreme values compared with a normal distribution.

There are three common types:

- **Mesokurtic** – Similar to a normal distribution.
- **Leptokurtic** – Heavier tails and more extreme values.
- **Platykurtic** – Lighter tails and fewer extreme values.

### Simple Example

Suppose we have two datasets:

- Dataset A has values mostly close to the average.
- Dataset B has some very large and very small values.

Dataset B has more extreme observations, so it can have **higher kurtosis**.

## Step 2 – Demonstrate the Concept

Kurtosis mainly helps us understand the **tails of a distribution**.

### Simple Representation

**Mesokurtic:**

       *
      ***
     *****
    *******
   *********
   
**Leptokurtic:**

        *
       ***
      *****
     ***
    *   *
   *     *

**Platykurtic:**

     *******
   ***********
  *************
 ***************

The important point is that kurtosis is mainly related to the **presence of extreme values and tail behaviour**, not simply how tall or short the graph looks.

### Real-World Examples

Kurtosis can be useful when studying:

- Financial returns
- Transaction values
- Measurement errors
- Customer spending
- Sensor data

```python
import numpy as np
import matplotlib.pyplot as plt

# Create a dataset
data = np.random.normal(loc=50, scale=10, size=1000)

# Calculate mean
mean = np.mean(data)

# Calculate the fourth standardized moment
kurtosis = np.mean(((data - mean) / np.std(data)) ** 4)

# Print the kurtosis value
print("Kurtosis:", kurtosis)

# Plot the distribution
plt.hist(data, bins=30)

# Add labels and title
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Kurtosis of a Normal Distribution")

# Display the graph
plt.show()
```

## Step 3 – Explanation

### What does the program do?

The program creates 1,000 values from a normal distribution.

It calculates the kurtosis using the fourth standardized moment and displays the distribution as a histogram.

### What statistical concept does it demonstrate?

It demonstrates **Kurtosis**, which describes the tail behaviour and presence of extreme values in a distribution.

For a normal distribution, this calculation gives a kurtosis close to **3**.

### How do we interpret the output?

The printed value will be approximately 3, although the exact value changes slightly because the dataset is randomly generated.

- Kurtosis ≈ 3 → Similar to a normal distribution.
- Kurtosis > 3 → Heavier tails and potentially more extreme values.
- Kurtosis < 3 → Lighter tails and fewer extreme values.

### Where is this used in AI/ML?

Kurtosis can be used during **Exploratory Data Analysis (EDA)** to identify features that may contain extreme observations or unusual tail behaviour.

It can help in:

- Detecting unusual data patterns
- Understanding feature distributions
- Identifying potential outliers
- Data preprocessing
- Financial and anomaly-detection datasets

# Empirical Rule (68–95–99.7)

## Step 1 – Understand the Concept

The **Empirical Rule**, also called the **68–95–99.7 Rule**, describes how data is distributed in a **normal distribution**.

For a normally distributed dataset:

- About **68%** of the data falls within **1 standard deviation** of the mean.
- About **95%** of the data falls within **2 standard deviations** of the mean.
- About **99.7%** of the data falls within **3 standard deviations** of the mean.

### Simple Example

Suppose:

- Mean = 100
- Standard Deviation = 10

Then:

- **68%** → between 90 and 110
- **95%** → between 80 and 120
- **99.7%** → between 70 and 130

This rule helps us quickly understand how much data is located around the mean.

## Step 2 – Demonstrate the Concept

The Empirical Rule can be visualized like this:

                  Mean
                   |
        68%        |
      <------------>
     -1 SD       +1 SD

    <------------------------>
             95%
    -2 SD                  +2 SD

<------------------------------------>
                 99.7%
-3 SD                              +3 SD

### Real-World Example

Suppose the average exam score is **70** and the standard deviation is **10**.

Then approximately:

- 68% of students score between **60 and 80**
- 95% score between **50 and 90**
- 99.7% score between **40 and 100**

This gives us a quick understanding of how scores are spread around the average.

```python
import numpy as np
import matplotlib.pyplot as plt

# Set the mean and standard deviation
mean = 100
std = 10

# Generate normally distributed data
data = np.random.normal(mean, std, 10000)

# Calculate the percentage within 1, 2, and 3 standard deviations
within_1_std = np.mean((data >= mean - std) & (data <= mean + std)) * 100
within_2_std = np.mean((data >= mean - 2 * std) & (data <= mean + 2 * std)) * 100
within_3_std = np.mean((data >= mean - 3 * std) & (data <= mean + 3 * std)) * 100

# Display the results
print(f"Within 1 standard deviation: {within_1_std:.2f}%")
print(f"Within 2 standard deviations: {within_2_std:.2f}%")
print(f"Within 3 standard deviations: {within_3_std:.2f}%")

# Plot the distribution
plt.hist(data, bins=50)

# Add labels and title
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Empirical Rule - 68-95-99.7")

# Display the graph
plt.show()
```

## Step 3 – Explanation

### What does the program do?

The program generates 10,000 values from a normal distribution.

It then calculates how many values fall within:

- 1 standard deviation from the mean
- 2 standard deviations from the mean
- 3 standard deviations from the mean

Finally, it displays the distribution using a histogram.

### What statistical concept does it demonstrate?

It demonstrates the **Empirical Rule (68–95–99.7 Rule)**.

For a normal distribution, approximately:

- 68% of values are within 1 standard deviation.
- 95% of values are within 2 standard deviations.
- 99.7% of values are within 3 standard deviations.

### How do we interpret the output?

Because the data is randomly generated, the percentages may not be exactly 68%, 95%, and 99.7%.

They should be close to these values.

For example:

- Within 1 SD → around 68%
- Within 2 SD → around 95%
- Within 3 SD → around 99.7%

### Where is this used in AI/ML?

The Empirical Rule is useful for:

- Understanding feature distributions
- Detecting unusual observations
- Identifying potential outliers
- Exploratory Data Analysis (EDA)
- Understanding standardized data

It is especially useful when a numerical feature approximately follows a normal distribution.
